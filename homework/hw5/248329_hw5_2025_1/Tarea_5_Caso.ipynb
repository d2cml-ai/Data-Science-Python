```python
# Instalar las bibliotecas necesarias
!pip install pymupdf
!pip install transformers
!pip install tqdm  # Instalar tqdm para la barra de progreso

# Importar las bibliotecas necesarias
import fitz  # PyMuPDF
from transformers import pipeline
from google.colab import files
from tqdm.notebook import tqdm  # Importar tqdm para barra de progreso

# Función para subir el archivo PDF
def upload_pdf():
    uploaded = files.upload()
    for filename in uploaded.keys():
        print(f'Archivo subido: {filename}')
        return filename

# Función para extraer el texto del PDF
def extract_text_from_pdf(pdf_path):
    doc = fitz.open(pdf_path)
    text = ""
    for page in doc:
        text += page.get_text()
    return text

# Función para dividir el texto en fragmentos más pequeños
def split_text(text, max_length=1000):
    words = text.split()
    chunks = []
    chunk = ""
    for word in words:
        if len(chunk) + len(word) + 1 <= max_length:
            chunk += " " + word
        else:
            chunks.append(chunk)
            chunk = word
    if chunk:
        chunks.append(chunk)
    return chunks

# Subir el archivo PDF
pdf_path = upload_pdf()

# Extraer el texto del archivo PDF subido
contract_text = extract_text_from_pdf(pdf_path)

# Dividir el texto extraído en fragmentos más pequeños
chunks = split_text(contract_text, max_length=1000)

# Cargar el modelo de Hugging Face para resumen
summarizer = pipeline("summarization", model="facebook/bart-large-cnn")

# Generar resúmenes para cada fragmento con barra de progreso
summaries = []
for chunk in tqdm(chunks, desc="Generando resúmenes", unit="fragmento"):
    summary = summarizer(chunk, max_length=300, min_length=100, do_sample=False)
    summaries.append(summary[0]['summary_text'])

# Combinar todos los resúmenes en un resumen final
final_summary = " ".join(summaries)

# Mostrar el resumen final
print("Resumen Ejecutivo:")
print(final_summary)
```
