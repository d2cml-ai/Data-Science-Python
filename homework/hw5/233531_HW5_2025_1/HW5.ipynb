```python
import pandas as pd

file_path = "FinancialPhraseBank-v1.0/Sentences_AllAgree.txt"
df = pd.read_csv(file_path, sep='@', names=["sentence", "label"], encoding="ISO-8859-1")

# Limpiar datos
df["sentence"] = df["sentence"].str.strip()
df["label"] = df["label"].str.strip().str.lower()

print(f"Número de frases en el dataset: {len(df)}")

df.head()
```

```python
from transformers import pipeline

# Cargar el modelo FinBERT especializado en finanzas
finbert = pipeline("sentiment-analysis", model="ProsusAI/finbert", tokenizer="ProsusAI/finbert")
```

```python
# Elegimos una muestra de 10 frases
muestra = df["sentence"][:10].tolist()

# Analizamos el sentimiento de cada frase
resultados = finbert(muestra)

# Mostrar resultados
for frase, r in zip(muestra, resultados):
    print(f"{frase}\n→ {r['label']} (confianza: {r['score']:.2f})\n")
```

```python
from google.colab import files

uploaded = files.upload()  # Elige tu archivo .pdf cuando te lo pida
```

```python
import fitz  # PyMuPDF

# Reemplaza por el nombre real
file_name = list(uploaded.keys())[0]

# Abrir el PDF
doc = fitz.open(file_name)

# Extraer texto de todas las páginas
texto = ""
for page in doc:
    texto += page.get_text()

# Mostrar parte del texto
print(texto[:1000])  # Solo muestra los primeros 1000 caracteres
```

```python
from transformers import pipeline

# Crear el pipeline de resumen
resumen = pipeline("summarization", model="facebook/bart-large-cnn")

# Usamos solo los primeros 2000 caracteres para empezar
texto_corto = texto[:2000]

# Generar el resumen
resultado = resumen(texto_corto, max_length=200, min_length=50, do_sample=False)

# Mostrar resumen
print("📝 RESUMEN DEL CONTRATO:")
print(resultado[0]['summary_text'])
```

```python
from transformers import pipeline

# 1. Crear el pipeline de resumen con BART
resumen = pipeline("summarization", model="facebook/bart-large-cnn")

# 2. Función para dividir el texto en fragmentos manejables (1000–1500 caracteres)
def dividir_texto(texto, max_len=1000):
    partes = []
    for i in range(0, len(texto), max_len):
        partes.append(texto[i:i+max_len])
    return partes

# 3. Dividimos el texto del contrato
fragmentos = dividir_texto(texto)

# 4. Generar resumen para cada fragmento
resumenes_parciales = []

for i, fragmento in enumerate(fragmentos):
    print(f"⏳ Resumiendo fragmento {i+1}/{len(fragmentos)}...")
    resumen_generado = resumen(fragmento, max_length=200, min_length=50, do_sample=False)
    resumenes_parciales.append(resumen_generado[0]['summary_text'])

# 5. Unir todos los resúmenes en uno solo
resumen_final = "\n\n".join(resumenes_parciales)

# 6. Mostrar el resumen completo
print("📄 RESUMEN EJECUTIVO DEL CONTRATO:")
print(resumen_final)
```
