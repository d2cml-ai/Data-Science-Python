```python
from datasets import load_dataset

# Cargar el Financial PhraseBank con permiso para ejecutar código remoto
dataset = load_dataset('financial_phrasebank', 'sentences_allagree', trust_remote_code=True)

# Ver los primeros registros del dataset
print(dataset['train'][0])
```

```python
from transformers import BertTokenizer, BertForSequenceClassification
from transformers import pipeline

# Cargar el modelo FinBERT y el tokenizador
model_name = 'yiyanghkust/finbert-tone'
tokenizer = BertTokenizer.from_pretrained(model_name)
model = BertForSequenceClassification.from_pretrained(model_name)

# Crear un pipeline de análisis de sentimientos
classifier = pipeline('sentiment-analysis', model=model, tokenizer=tokenizer)
```

```python
# Ver los primeros 3 ejemplos del dataset para verificar la estructura
print(dataset['train'][:3])
```

```python
# Aplicar el clasificador a las primeras 10 frases del dataset
predictions = [classifier(sentence)[0]['label'] for sentence in dataset['train']['sentence'][:10]]

# Mostrar los resultados
for sentence, sentiment in zip(dataset['train']['sentence'][:10], predictions):
    print(f"Frase: {sentence} - Sentimiento: {sentiment}")
```

```python
import pandas as pd

# Aplicar el clasificador a todas las frases del dataset
predictions = [classifier(sentence)[0]['label'] for sentence in dataset['train']['sentence']]

# Crear un DataFrame con todas las frases y sus sentimientos
results = pd.DataFrame({
    'sentence': dataset['train']['sentence'],  # Todas las frases
    'predicted_sentiment': predictions  # Todas las predicciones
})

# Guardar el DataFrame a un archivo CSV
results.to_csv('sentiment_predictions_all.csv', index=False)

print("Resultados guardados en 'sentiment_predictions_all.csv'")
```
