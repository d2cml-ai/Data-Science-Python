## Let's get started

```python
import numpy as np
import pandas as pd
import os
import time
import re
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.support.ui import Select
from selenium.webdriver.chrome.service import Service
from selenium.common.exceptions import NoSuchElementException, StaleElementReferenceException
from selenium.webdriver import ActionChains
```

### We apply the filters for the job offers

```python
# Initialize the Chrome WebDriver
driver = webdriver.Chrome()

# URL of the job listings page
url = "https://www.bumeran.com.pe/empleos.html"

# Open the page
driver.get(url)

# Wait for 3 seconds to allow the page to load
time.sleep(3)

# Maximize the browser window
driver.maximize_window()

# Wait for 2 seconds to ensure the window is maximized
time.sleep(2)

# Find the "Publication Date" button and click it
publication_date = driver.find_element(By.XPATH, '//button[contains(text(), "Fecha de publicación")]')
publication_date.click()

# Wait for 2 seconds
time.sleep(2)

# Find the "Less than 15 days" button and click it
less_than_15_days = driver.find_element(By.XPATH, '//button[contains(text(), "Menor a 15 días")]')
less_than_15_days.click()

# Wait for 2 seconds
time.sleep(2)

# Find the "Area" button and click it
area_button = WebDriverWait(driver, 10).until(
    EC.element_to_be_clickable((By.XPATH, "//button[contains(text(), 'Área')]")))
area_button.click()

# Wait for 2 seconds
time.sleep(2)

# Find the "Technology, Systems and Telecommunications" button and click it
technology_button = WebDriverWait(driver, 10).until(
    EC.element_to_be_clickable((By.XPATH, "//button[contains(text(), 'Tecnología, Sistemas y Telecomunicaciones')]")))
technology_button.click()

# Find the "Subarea" button and click it
subarea_button = WebDriverWait(driver, 10).until(
    EC.element_to_be_clickable((By.XPATH, "//button[contains(text(), 'Subárea')]")))
subarea_button.click()

# Wait for 2 seconds
time.sleep(2)

# Find the "Programming" button and click it
programming_button = WebDriverWait(driver, 10).until(
    EC.element_to_be_clickable((By.XPATH, "//button[contains(text(), 'Programación')]")))
programming_button.click()

# Wait for 2 seconds
time.sleep(2)

# Find the "Department" button and click it
department_button = WebDriverWait(driver, 10).until(
    EC.element_to_be_clickable((By.XPATH, "//button[contains(text(), 'Departamento')]")))
department_button.click()

# Wait for 2 seconds
time.sleep(2)

# Find the "Lima" button and click it
lima_button = WebDriverWait(driver, 10).until(
    EC.element_to_be_clickable((By.XPATH, "//button[contains(text(), 'Lima')]")))
lima_button.click()

# Wait for 2 seconds
time.sleep(2)

# Find the "Workload" button and click it
workload_button = WebDriverWait(driver, 10).until(
    EC.element_to_be_clickable((By.XPATH, "//button[contains(text(), 'Carga horaria')]")))
workload_button.click()

# Wait for 2 seconds
time.sleep(2)

# Find the "Full-time" button and click it
full_time_button = WebDriverWait(driver, 10).until(
    EC.element_to_be_clickable((By.XPATH, "//button[contains(text(), 'Full-time')]")))
full_time_button.click()

# Wait for 2 seconds
time.sleep(2)
```

# STAGE 1 

 We will extract all the job offer links

```python
# List to store job offer links
job_links = []

# Loop to iterate over the different result pages
while True:
    # Wait for the main results section to be fully loaded
    WebDriverWait(driver, 10).until(
        EC.presence_of_element_located((By.ID, "listado-avisos"))
    )
    
    # Find the container that holds the job offer links
    results_section = driver.find_element(By.ID, "listado-avisos")
    job_elements = results_section.find_elements(By.TAG_NAME, "a")
    
    # Filter the relevant job offer links
    for element in job_elements:
        job_url = element.get_attribute("href")
        if job_url and "/empleos/" in job_url and job_url not in job_links:
            job_links.append(job_url)
    
    # Print the number of collected links so far
    print(f" Total links accumulated: {len(job_links)}")

    # Try to click on the next page arrow
    try:
        next_button = driver.find_element(By.XPATH, '//*[@id="listado-avisos"]/div[22]/a[2]/i')
        driver.execute_script("arguments[0].click();", next_button)
        time.sleep(5)  # Wait a little for the new page to load
    except:
        # If the arrow is not found, it means there are no more pages
        print("No more pages available.")
        break

# Print the final list of collected links
print("\n Collected links:")
for link in job_links:
    print(link)
```

# STAGE: 2 

We will extract the job titles, job descriptions, district, and work modality, and combine them into a dataframe along with their links.


```python
# Now we navigate through the first 10 collected job links
jobs_data = []  # List to store job information

# We limited it to the first 10 links because otherwise it would take quite a long time. 
#If you need all the information, you can modify the code accordingly
for idx, url in enumerate(job_links[:10], start=1):
    driver.get(url)
    time.sleep(5)  # Wait for the page to load

    try:
        # Job title
        title = driver.find_element(By.TAG_NAME, "h1").text

        # Description
        container = driver.find_element(By.ID, "ficha-detalle")
        description = container.text.strip()

        # District (Location)
        district = driver.find_element(
            By.XPATH, '//*[@id="ficha-detalle"]/div[2]/div/div[1]/div[1]/div[2]/div/div/li/a/h2'
        ).text

        # Modality (Remote, On-site, Hybrid)
        try:
            WebDriverWait(driver, 5).until(
                EC.presence_of_all_elements_located((By.TAG_NAME, "p"))
            )
            modality_tags = driver.find_elements(By.TAG_NAME, "p")
            modality = next(
                (tag.text.strip() for tag in modality_tags if tag.text.strip() in ["Remoto", "Presencial", "Híbrido"]),
                "NO MODALITY"
            )
        except:
            modality = "NO MODALITY"

        # Store the information in the list
        jobs_data.append({
            "Link": url,
            "Title": title,
            "Description": description,
            "Modality": modality,
            "District": district,
        })

        # Progress output
        print(f"[{idx}/10] Job processed: {title}")

    except Exception as e:
        print(f"Error while processing URL {url}: {e}")

# Create a DataFrame with the collected data
df = pd.DataFrame(jobs_data)

# Display the DataFrame
print("\nFinal collected data:")

```
