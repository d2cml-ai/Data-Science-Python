```python
# Descargar archivo de geocode

# 0. Bibliotecas. Las pongo acá porque sino parece como si nunca las hubiese descargado.
!pip install pandas as pd
import pandas as pd
import osmnx as ox
import geopandas as gpd
import time

# Lista de regiones grandes de Perú (puedes agregar más si quieres más precisión)
departamentos = [
    "Loreto, Peru", "Ucayali, Peru", "Madre de Dios, Peru", 
    "San Martín, Peru", "Cusco, Peru", "Arequipa, Peru", 
    "Lima, Peru", "Piura, Peru", "Puno, Peru"
]

# Tags requeridos
tags = {
    'boundary': 'protected_area',
    'leisure': 'nature_reserve',
    'protect_class': True
}

# DataFrame vacío para guardar todos los resultados
gdf_total = gpd.GeoDataFrame()

# Bucle con barra de progreso simulada
for i, region in enumerate(departamentos, 1):
    print(f"[{i}/{len(departamentos)}] Descargando datos para: {region}...")
    try:
        poly = ox.geocode_to_gdf(region).geometry.iloc[0]
        gdf = ox.features_from_polygon(poly, tags=tags)
        gdf_total = pd.concat([gdf_total, gdf], ignore_index=True)
        print(f"✔️ {region} completado. Total acumulado: {len(gdf_total)} registros\n")
    except Exception as e:
        print(f"❌ Error en {region}: {e}\n")
    time.sleep(1)  # para no saturar Overpass API

# Guardar todo al final
gdf_total.to_file("peru_protected_areas.gpkg", driver="GPKG")
print("✅ Archivo guardado como peru_protected_areas.gpkg")
```
