```python
from pypdf import PdfReader
```

The PdfReader class stores the information for each page in the PDF document we have loaded. They are contained within the pages attribute, an iterable.





```python
from google.colab import files
from pypdf import PdfReader

# Upload the PDF
uploaded = files.upload()

# Function to extract text from the PDF using pypdf
def extract_text_from_pdf(pdf_path):
    reader = PdfReader(pdf_path)
    full_text = ""
    for page in reader.pages:
        full_text += page.extract_text()
    return full_text

# Get the file name of the uploaded PDF
pdf_file_path = list(uploaded.keys())[0]

# Extract text from the uploaded PDF
extracted_text = extract_text_from_pdf(pdf_file_path)

# Print the first 1000 characters to verify the extracted text
print(extracted_text[:1000])
```

```python
def pdfToString(path):
    documentText = ""
    loadedPdf = PdfReader(path)
    for page in loadedPdf.pages:
        documentText += page.extract_text()
    return documentText
```

```python
from openai import OpenAI
from getpass import getpass

# Pedir clave y limpiar espacios
api_key = getpass("Ingresa tu API key de OpenAI: ").strip()

# Crear el cliente
client = OpenAI(api_key=api_key)

# Probar la conexión
response = client.chat.completions.create(
    model="gpt-3.5-turbo",
    messages=[{"role": "user", "content": "¿Estás funcionando?"}]
)

print(response.choices[0].message.content)
```

```python
from textwrap import dedent

# 3. Definir el prompt con textwrap.dedent para evitar indentaciones indeseadas
prompt_template = dedent("""
    Analyze the following text and provide a ranking of the urban parks in Northern Lima
    based on their efficiency in carbon dioxide capture and tree cover.
    Return the parks from most to least efficient and give a brief justification for each.

    Text:
    {context}
""")

combined_prompt = prompt_template.format(context=extracted_text)

# 4. Llamar al modelo
try:
    response = client.chat.completions.create(
        model="gpt-3.5-turbo",
        messages=[
            {"role": "system", "content": "You are a helpful assistant."},
            {"role": "user", "content": combined_prompt}
        ],
        max_tokens=1_500,
        temperature=0.7,
        # timeout y retries heredados del cliente; se pueden ajustar con .with_options()
    )

    print(response.choices[0].message.content)

except openai.APIError as e:           # Errores de la API (4xx/5xx)
    print(f"OpenAI API error: {e}")
except openai.APIConnectionError as e: # Problemas de red / timeouts
    print(f"Connection error: {e}")
except Exception as e:                 # Cualquier otro error
    print(f"Unexpected error: {e}")
```

### Sería el mismo procedimiento de los prompts para optimal species recommendations, and district-level diagnostics.

```python
from textwrap import dedent

species_prompt = dedent("""
    Analyze the following text and provide a optimal species recommendations.

    Text:
    {context}
""")

combined_species_prompt = species_prompt.format(context=extracted_text)

try:
    response_species = client.chat.completions.create(
        model="gpt-3.5-turbo",
        messages=[
            {"role": "system", "content": "You are a helpful and expert assistant."},
            {"role": "user", "content": combined_species_prompt}
        ],
        max_tokens=1500,
        temperature=0.7,
    )

    print("\n🔎 Optimal Species Recommendations:\n")
    print(response_species.choices[0].message.content)

except Exception as e:
    print(f"Error during species recommendation request: {e}")
```

```python
district_prompt = dedent("""
    Analyze the following urban park report and provide a diagnostic summary by district in Northern Lima.
    For each district, summarize the key findings regarding:
      - Carbon dioxide capture efficiency
      - Tree cover density
      - Notable strengths or weaknesses

    Present the output clearly structured per district.

    Text:
    {context}
""")

combined_district_prompt = district_prompt.format(context=extracted_text)

try:
    response_district = client.chat.completions.create(
        model="gpt-3.5-turbo",
        messages=[
            {"role": "system", "content": "You are an urban ecology analyst."},
            {"role": "user", "content": combined_district_prompt}
        ],
        max_tokens=1500,
        temperature=0.7,
    )

    print("\n📍 District-Level Diagnostics:\n")
    print(response_district.choices[0].message.content)

except Exception as e:
    print(f"Error during district diagnostics request: {e}")
```

```python
from textwrap import dedent

policy_prompt = dedent("""
    Based on the following information about urban parks in Northern Lima,
    assess how the data on carbon dioxide capture and tree cover
    can be used to inform and support public policy development.

    Consider policy areas such as:
      - Urban planning and green infrastructure
      - Environmental and climate policy
      - Health and social equity
      - Municipal resource allocation

    Provide concrete examples of policy recommendations that could be derived from the findings.

    Text:
    {context}
""")

combined_policy_prompt = policy_prompt.format(context=extracted_text)

try:
    response_policy = client.chat.completions.create(
        model="gpt-3.5-turbo",
        messages=[
            {"role": "system", "content": "You are a public policy advisor with expertise in urban sustainability."},
            {"role": "user", "content": combined_policy_prompt}
        ],
        max_tokens=1500,
        temperature=0.7,
    )

    print("\n🏛️ Policy Development Recommendations:\n")
    print(response_policy.choices[0].message.content)

except Exception as e:
    print(f"Error during policy recommendation request: {e}")
```
