```python
from selenium import webdriver
import re
import time
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By

#Driver
driver = webdriver.Chrome()
#Ingresar pagina web, ajustar pantalla y zoom
url = 'https://www.bumeran.com.pe/empleos.html'
driver.get(url)

options = webdriver.ChromeOptions()
driver = webdriver.Chrome(options=options)
driver.maximize_window()
driver.execute_script("document.body.style.zoom='100%'")
time.sleep(3)

#Aplicar filtros
#Fecha
driver.find_element(By.XPATH, "//button[contains(text(), 'Fecha')]").click()
time.sleep(3)
driver.find_element(By.XPATH, "//button[contains(text(), 'Menor a 15 días')]").click()
time.sleep(3)
#Area
driver.find_element(By.XPATH, "//button[contains(text(), 'Área')]").click()
time.sleep(3)
driver.find_element(By.XPATH, "//button[contains(text(), 'Tecnología')]").click()
time.sleep(3)
#Subarea
driver.find_element(By.XPATH, "//button[contains(text(), 'Subárea')]").click()
time.sleep(3)
driver.find_element(By.XPATH, "//button[contains(text(), 'Programación')]").click()
time.sleep(3)
#Dpto
driver.find_element(By.XPATH, "//button[contains(text(), 'Departamento')]").click()
time.sleep(3)
driver.find_element(By.XPATH, "//button[contains(text(), 'Lima')]").click()
time.sleep(3)
#Horario
driver.find_element(By.XPATH, "//button[contains(text(), 'Carga horaria')]").click()
time.sleep(3)
driver.find_element(By.XPATH, "//button[contains(text(), 'Full-time')]").click()
time.sleep(3)

#Encontrar todos los urls de ofertas con href
ofertas = driver.find_elements(By.XPATH, "//a[contains(@href, '/empleos/')]")
#Extraer urls y guardarlos en lista urls_ofertas
urls_ofertas = [oferta.get_attribute("href") for oferta in ofertas]
#Mostrar todos los urls
print(urls_ofertas)

import csv

#Lista para almacenar los datos
datos_ofertas = []

# Extraer datos de cada oferta
for url in urls_ofertas:
    driver.get(url)
    time.sleep(5)
    try:
        #Extraer título
        titulo = driver.find_element(By.TAG_NAME, "h1").text
        
        #Extraer descripción hasta Beneficios
        descripcion_completa = driver.find_element(By.ID, "ficha-detalle").text
        descripcion = descripcion_completa.split("Beneficios")[0] if "Beneficios" in descripcion_completa else descripcion_completa
        
        #Extraer distrito
        distrito = driver.find_element(By.XPATH, "//div[contains(@class, 'ubicacion')]//h2").text
       
        #Extraer modalidad
        if "presencial" in descripcion_completa or "Presencial" in descripcion_completa: modalidad = "Presencial",
        elif "híbrido" in descripcion_completa or "hibrido" in descripcion_completa or "Híbrido" in descripcion_completa or "Hibrido" in descripcion_completa: modalidad = "Híbrido",
        elif "remoto" in descripcion_completa or "virtual" in descripcion_completa or "Remoto" in descripcion_completa or "Virtual" in descripcion_completa: modalidad = "Remoto",
        else: modalidad = "No especificado"
        
        #Agregar datos a la lista
        datos_ofertas.append([titulo, descripcion, distrito, modalidad, url])
   
    except Exception as e:
        print(f"Error en {url}: {e}")

# Guardar en CSV
with open("ofertas_trabajo.csv", "w", newline="", encoding="utf-8") as archivo:
    escritor = csv.writer(archivo)
    escritor.writerow(["Título", "Descripción", "Distrito", "Modalidad", "URL"])
    escritor.writerows(datos_ofertas)

print("Datos guardados en ofertas_trabajo.csv")
driver.quit()
```
