```python
# Libreria para manejar paginas
from selenium import webdriver


import re
import time 
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import pandas as pd
```

```python
#el driver
driver = webdriver.Chrome()
#la pagina
url = "https://www.bumeran.com.pe/empleos.html"
driver.get(url)
driver.maximize_window()
time.sleep(3)
driver.execute_script("document.body.style.zoom='100%'")
time.sleep(3)
#colocando filtros
fecha = driver.find_element(By.XPATH,"/html/body/div[1]/div/div[2]/div[1]/div/div[2]/div[1]/div[2]/div/div[1]/button[1]")
fecha.click()
time.sleep(3)
dias_15 = driver.find_element(By.XPATH,"//button[contains(text(), 'Menor a 15 días')]")
dias_15.click()
time.sleep(3)
area = driver.find_element(By.XPATH,"/html/body/div[1]/div/div[2]/div[1]/div/div[2]/div[1]/div[2]/div/div[1]/button[1]")
area.click()
time.sleep(3)
tecno = driver.find_element(By.XPATH,"/html/body/div[1]/div/div[2]/div[1]/div/div[2]/div[1]/div[2]/div/div[2]/div/div/button[3]")
tecno.click()
time.sleep(3)
subarea = driver.find_element(By.XPATH,"/html/body/div[1]/div/div[2]/div[1]/div/div[2]/div[1]/div[2]/div/div[1]/button[1]")
subarea.click()
time.sleep(3)
program = driver.find_element(By.XPATH,"/html/body/div[1]/div/div[2]/div[1]/div/div[2]/div[1]/div[2]/div/div[2]/div/div/button[1]")
program.click()
time.sleep(3)
depa = driver.find_element(By.XPATH,"/html/body/div[1]/div/div[2]/div[1]/div/div[2]/div[1]/div[2]/div/div[1]/button[1]")
depa.click()
time.sleep(3)
lima = driver.find_element(By.XPATH,"/html/body/div[1]/div/div[2]/div[1]/div/div[2]/div[1]/div[2]/div/div[2]/div/div/button[1]")
lima.click()
time.sleep(3)
carga = driver.find_element(By.XPATH,"/html/body/div[1]/div/div[2]/div[1]/div/div[2]/div[1]/div[2]/div/div[1]/button[4]")
carga.click()
time.sleep(3)
full = driver.find_element(By.XPATH,"/html/body/div[1]/div/div[2]/div[1]/div/div[2]/div[1]/div[2]/div/div[2]/div/div/button[1]")
full.click()
time.sleep(3)


all_links = []  # Lista para guardar todas las URLs

while True:
    # Esperar a que carguen los empleos
    WebDriverWait(driver, 10).until(
        EC.presence_of_all_elements_located((By.CLASS_NAME, "sc-zDqdV"))
    )

    # Extraer los enlaces de empleos en la página actual
    empleos = driver.find_elements(By.CLASS_NAME, "sc-zDqdV")
    links = [job.get_attribute("href") for job in empleos]

    # Agregar los links nuevos a la lista general
    all_links.extend(links)

    print(f"Extraídos {len(links)} empleos en esta página.")

    # Intentar encontrar el botón "Siguiente"
    try:
        siguiente = driver.find_element(By.XPATH, "/html/body/div[1]/div/div[2]/div[1]/div/div[2]/div[2]/div[22]/a[2]/i")

        # Si el botón está deshabilitado, salimos del bucle
        if "disabled" in siguiente.get_attribute("class"):
            break

        # Hacer clic en el botón "Siguiente"
        siguiente.click()
        time.sleep(3)  # Esperar a que cargue la siguiente página

    except:
        print("No hay más páginas.")
        break  # Si no encuentra el botón, termina el ciclo


  
  # Guardar en un archivo Excel porsiacaso
df = pd.DataFrame({"URL": all_links})
df.to_excel("ofertas_laborales.xlsx", index=False)

print("URLs extraídas y guardadas en ofertas_laborales.xlsx")
print(all_links)
# Cerrar el navegador
driver.quit()

```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
import pandas as pd

def get_distrito(driver):
    
    xpaths = ["/html/body/div[1]/div/div[2]/div[2]/div/div[2]/div[1]/div[2]/div[1]/div[1]/div[2]/div/div/div[1]/div[2]/div/div/li/a/h2",   
        "/html/body/div[1]/div/div[2]/div[2]/div/div[2]/div[1]/div[2]/div[1]/div[1]/div[2]/div/div[1]/div[1]/div[2]/div/div/li/a/h2",  
    ]
    
    for xpath in xpaths:
        try:
            distrito = WebDriverWait(driver, 5).until(
                EC.presence_of_element_located((By.XPATH, xpath))
            ).text
            return distrito  
        except:
            continue  
    
    return "Distrito no encontrado"  # Si ninguno funciona
def get_modo_trabajo(driver):
    
    xpaths = [
        "/html/body/div[1]/div/div[2]/div[2]/div/div[2]/div[1]/div[2]/div[1]/div[1]/div[2]/div/div[1]/div[4]/div/ul/div[1]/li[1]/a/p",  
        "/html/body/div[1]/div/div[2]/div[2]/div/div[2]/div[1]/div[2]/div[1]/div[1]/div[2]/div/div/div[4]/div/ul/div[1]/li[1]/a/p",
        "/html/body/div[1]/div/div[2]/div[2]/div/div[2]/div[1]/div[2]/div[1]/div[1]/div[2]/div/div/div[5]/div/ul/div[1]/li[1]/a/p",
        "/html/body/div[1]/div/div[2]/div[2]/div/div[2]/div[1]/div[2]/div[1]/div[1]/div[2]/div/div[1]/div[5]/div/ul/div[1]/li[1]/a/p",
          
    ]

    for xpath in xpaths:
        try:
            modo = WebDriverWait(driver, 5).until(
                EC.presence_of_element_located((By.XPATH, xpath))
            ).text
            return modo  
        except:
            continue 

    return "Modo de trabajo no encontrado"  

driver = webdriver.Chrome()
info_empleos = []

for url in all_links:
    time.sleep(3)
    driver.get(url)
    driver.maximize_window()
    time.sleep(3)

    try:
        # Extraer el título del empleo
        titulo = WebDriverWait(driver, 10).until(
            EC.presence_of_element_located((By.XPATH, "/html/body/div[1]/div/div[2]/div[2]/div/div[1]/div/div/div[1]/div/div[1]/h1"))  
        ).text

        # Extraer la descripción 
        descripcion = WebDriverWait(driver, 10).until(
            EC.presence_of_element_located((By.XPATH, "/html/body/div[1]/div/div[2]/div[2]/div/div[2]/div[1]/div[2]/div[1]"))).text 

        # Extraer el distrito 
        distrito = get_distrito(driver)

        # Extraer el modo de trabajo 
        modo = get_modo_trabajo(driver)

        # Guardar los datos en la lista
        info_empleos.append({
            "Empleo": titulo,
            "Descripcion": descripcion,
            "Distrito": distrito,
            "Modo de trabajo": modo,
            "URL": url
        })

        print(f"Extraído: {titulo}")

    except Exception as e:
        print(f"Error en {url}: {e}")
        continue  # Si falla un empleo, pasa al siguiente

# Guardar los datos en un nuevo archivo Excel
df_jobs = pd.DataFrame(info_empleos)
df_jobs.to_excel("Empleos_programacion.xlsx", index=False)
df_jobs.to_csv("Empleos_programacioncsv.csv", index=False, encoding="utf-8")

print("Datos extraídos y guardados en 'Empleos_programacion.xlsx'.")

# Cerrar el navegador
driver.quit()
```
