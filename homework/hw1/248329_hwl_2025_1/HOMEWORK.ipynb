```python
# Inicializar el WebDriver
driver = webdriver.Chrome()
driver.maximize_window()

# URL
url = 'https://www.bumeran.com.pe/en-lima/empleos-area-tecnologia-sistemas-y-telecomunicaciones-subarea-programacion-full-time-publicacion-menor-a-15-dias.html'
driver.get(url)
```

```python
job_urls = []  # Lista para almacenar los enlaces de empleos

while True:
    # Esperar un poco para cargar la página
    time.sleep(3)

    # Encuentra todos los enlaces de empleos en la página actual
    job_links = driver.find_elements(By.CSS_SELECTOR, "div.sc-kXoVnq a.sc-zDqdV")

    # Extraer y guardar los links
    for job in job_links:
        link = job.get_attribute('href')
        if link not in job_urls:
            job_urls.append(link)

    # Intentar encontrar el botón "Siguiente" y hacer clic si existe
    try:
        # Buscamos el botón "Siguiente" que tiene la clase que nos indicaste
        next_button = driver.find_element(By.CSS_SELECTOR, "a.sc-wENpt.hWWVNo")
        
        # Si existe el botón, hacemos clic
        if next_button.is_enabled():  # Asegurarse de que el botón esté habilitado
            next_button.click()
            time.sleep(3)  # Esperar a que cargue la siguiente página
        else:
            print("No hay más páginas. Fin del scraping.")
            break  # Salir del bucle si no hay más páginas
    except:
        print("No hay más páginas. Fin del scraping.")
        break  # Salir del bucle si no se encuentra el botón


# Imprimir todos los enlaces obtenidos
print("Enlaces de empleos encontrados:")
print(job_urls)
```

```python
print(len(job_urls))
```

```python
# Ahora navegamos por cada uno de los enlaces recopilados
jobs_data = []  # Lista para almacenar la información de los trabajos

# Recorrer todos los enlaces de trabajos
for url in job_urls:
    driver.get(url)
    time.sleep(5)  # Esperar a que la página cargue

    try:
        # Título del trabajo
        title = driver.find_element(By.TAG_NAME, "h1").text

        # Descripción
        description = driver.find_element(By.TAG_NAME, "p").text

        # Distrito (Ubicación)
        district = driver.find_element(By.XPATH, '//*[@id="ficha-detalle"]/div[2]/div/div[1]/div[1]/div[2]/div/div/li/a/h2').text

        # Almacenar la información en la lista
        jobs_data.append({
            "Link": url,
            "Título": title,
            "Descripción": description,
            "Distrito": district
        })

    except Exception as e:
        print(f"Error al procesar la URL {url}: {e}")

# Crear un DataFrame con los datos recopilados
df = pd.DataFrame(jobs_data)


# Mostrar el DataFrame
print(df)
```

```python
def asignar_work_mode(descripcion):
    if 'Híbrido' in descripcion:
        return 'Híbrido'
    elif 'Presencial' in descripcion:
        return 'Presencial'
    elif 'Remoto' in descripcion:
        return 'Remoto'
    else:
        return 'Otro'

# Aplicar la función a la columna "Descripción" y crear la nueva columna "Work_Mode"
df['Work_Mode'] = df['Descripción'].apply(asignar_work_mode)

# Mostrar el DataFrame resultante
print(df)
```
