## Las librerias necesarias:


```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
import pandas as pd
import re
```

```python
# Configurar Selenium
driver = webdriver.Chrome()
driver.maximize_window()

# Ir a la página web
url = "https://www.bumeran.com.pe/empleos.html"  # Reemplaza con la URL correcta
driver.get(url)
#--------------Los filtros de búsqueda------------------


#--------------Fecha de publicación-------------

fecha_publicacion = WebDriverWait(driver, 10).until(
    EC.element_to_be_clickable((By.XPATH, "//button[contains(@class, 'sc-kwgWOH') and contains(., 'Fecha de publicación')]"))
)
fecha_publicacion.click()


menor_15_dias = WebDriverWait(driver, 10).until(
    EC.element_to_be_clickable((By.XPATH, "//button[contains(@class, 'sc-ktBuXk') and contains(., 'Menor a 15 días')]"))
)
menor_15_dias.click()


#--------------Area-------------
area_boton = WebDriverWait(driver, 10).until(
    EC.element_to_be_clickable((By.XPATH, "//button[contains(@class, 'sc-kwgWOH') and contains(., 'Área')]"))
)
area_boton.click()

tecnologia_boton = WebDriverWait(driver, 10).until(
    EC.element_to_be_clickable((By.XPATH, "//button[contains(@class, 'sc-ktBuXk') and contains(., 'Tecnología, Sistemas y Telecomunicaciones')]"))
)
tecnologia_boton.click()

#--------------Sub Area-------------
subarea_boton = WebDriverWait(driver, 10).until(
    EC.element_to_be_clickable((By.XPATH, "//button[contains(@class, 'sc-kwgWOH') and contains(., 'Subárea')]"))
)
subarea_boton.click()

programacion_boton = WebDriverWait(driver, 10).until(
    EC.element_to_be_clickable((By.XPATH, "//button[contains(@class, 'sc-ktBuXk') and contains(., 'Programación')]"))
)
programacion_boton.click()


#--------------Departamento-------------
departamento = WebDriverWait(driver, 10).until(
    EC.element_to_be_clickable((By.XPATH, "//button[contains(@class, 'sc-kwgWOH') and contains(., 'Departamento')]"))
)
departamento.click()
lima_boton = WebDriverWait(driver, 10).until(
    EC.element_to_be_clickable((By.XPATH, "//button[contains(@class, 'sc-ktBuXk') and contains(., 'Lima')]"))
)
lima_boton.click()

#--------------Tipo de contrato-------------
carga_horaria = WebDriverWait(driver, 10).until(
    EC.element_to_be_clickable((By.XPATH, "//button[contains(@class, 'sc-kwgWOH') and contains(., 'Carga horaria')]"))
)
#driver.execute_script("arguments[0].scrollIntoView();", carga_horaria)
carga_horaria.click()
full_time = WebDriverWait(driver, 10).until(
    EC.element_to_be_clickable((By.XPATH, "//button[contains(@class, 'sc-ktBuXk') and contains(., 'Full-time')]"))
)
full_time.click()
```

## Recopilar los Links de cada página

```python
# Esperar a que los elementos de paginación carguen
WebDriverWait(driver, 10).until(
    EC.presence_of_element_located((By.CLASS_NAME, "sc-cGDfzg"))
)

# Obtener todas las páginas disponibles en la paginación
paginas = driver.find_elements(By.XPATH, "//div[contains(@class, 'sc-cGDfzg')]/a")

# Extraer las referencias de las páginas
links_paginas = [pagina.get_attribute("href") for pagina in paginas if pagina.get_attribute("href")]

print(f"Total de páginas: {len(links_paginas)}")
for i, link in enumerate(links_paginas, start=1):
    print(f"Página {i}: {link}")
```

```python

url_empleos = []
for i, link in enumerate(links_paginas, start=1):
    print(f"\nPágina {i}: {link}")
    driver.get(link)
    time.sleep(5)  # Espera inicial

    try:
        # Esperar a que cargue el listado
        WebDriverWait(driver, 15).until(
            EC.presence_of_element_located((By.ID, "listado-avisos"))
        )

        # Obtener ENLACES de empleos (usando href como filtro)
        empleos = driver.find_elements(
            By.XPATH, 
            '//div[@id="listado-avisos"]//a[contains(@href, "/empleos/")]'
        )

        # Extraer URLs válidas
        links_empleos = [empleo.get_attribute("href") for empleo in empleos if empleo.get_attribute("href")]
        print(f"Encontrados: {len(links_empleos)} empleos")
        url_empleos.extend(links_empleos)  # Usar extend para lista plana como cantidad de empleos

    except Exception as e:
        print(f"Error: {str(e)}")

# Cerrar el navegador
driver.quit()
print(f"Total de empleos encontrados: {len(url_empleos)}")        
```

```python
# Lista de URLs de páginas
links_paginas = url_empleos[:30]  # Limitar a las primeras 30 páginas para la prueba
# Configurar Selenium
driver = webdriver.Chrome()
driver.maximize_window()

DATA= []
for i, link in enumerate(links_paginas, start=1):
    daT=[]  # Inicializar la lista para cada página
    driver.get(link)
    time.sleep(5)  # Espera inicial

    try:
        WebDriverWait(driver, 15).until(
            EC.presence_of_element_located((By.ID, "ficha-detalle"))
        )
        
        #----------Distrito----------
        distrito = WebDriverWait(driver, 15).until(
            EC.visibility_of_element_located(
                (By.CSS_SELECTOR, '#ficha-detalle > div:nth-child(2) > div > div.sc-gWIwhj.zkMuK > div:nth-child(1) > div:nth-child(2) > div > div > li > a > h2')
            )
        ).text

        distrito = distrito.split(",")[0].strip()  # Limpiar el texto del distrito



        # ----------Título del trabajo---------- 
        job_title = WebDriverWait(driver, 15).until(
            EC.visibility_of_element_located(
                (By.CSS_SELECTOR, '#header-component > div.sc-giPFmd.eAXioh > div > div.sc-doIlLK.iMoClT > h1')
            )
        ).text
        
        
        # ----------Descripción del trabajo---------- (Una musa)

        # Buscar directamente el encabezado "Descripción del puesto"
        descripcion_header = WebDriverWait(driver, 10).until(
            EC.visibility_of_element_located((
                By.XPATH, "//h3[contains(text(), 'Descripción del puesto')]"
            ))
        )
        # Navegar al div padre
        div_padre = descripcion_header.find_element(By.XPATH, "./..")

        # Obtener el siguiente elemento hermano del div padre que en este caso contiene la descripción completa en </p>
        descripcion_contenedor = div_padre.find_element(By.XPATH, "following-sibling::*[1]")
        
        # Extraer todos los párrafos y listas para obtener el contenido completo
        elementos_texto = descripcion_contenedor.find_elements(By.XPATH, './/p|.//ul/li')
        
        # Unir todos los textos preservando estructura
        texto_completo = []
        for elem in elementos_texto:
            texto = elem.text.strip()
            if texto:  # Solo agregar si no está vacío
                texto_completo.append(texto)
        
        # Unir con saltos de línea con " " para preservar estructura cuando se guarde en CSV
        #tmbien se puede usar "\n" para separar los elementos, pero opte por " " para evitar saltos de línea innecesarios
        descripcion_final = " ".join(texto_completo)  

        # Unir todos los textos preservando estructura de la tarea 

        daT.extend([job_title])
        daT.extend([descripcion_final])
        daT.extend([distrito])

        DATA.append(daT)  
          
    except Exception as e:
        print(f"Error en la pagina{i}: {link}")
       
# Cerrar el navegador
driver.quit()
print(f"Total de datos encontrados: {len(DATA)}")
#Si no se encuentra la descripcion, agregar timepo de espera o revisar las páginas
```

## Buscar la modalidad : ) 

```python
import re


driver = webdriver.Chrome()
driver.maximize_window()


resultados = []

for url in links_paginas:
    try:
        driver.get(url)
        time.sleep(3)  # Breve espera para carga
        
        modalidad_encontrada = None
        
        # Buscar el elemento h2 con la ruta de navegación
        try:
            # Buscar todos los elementos h2 que puedan contener la ruta de navegación
            elementos_h2 = driver.find_elements(By.TAG_NAME, "h2")
            
            for h2 in elementos_h2:
                # Verificar si contiene enlaces
                enlaces = h2.find_elements(By.TAG_NAME, "a")
                
                if enlaces:
                    # Revisar cada enlace por el patrón modalidad-xxx
                    for enlace in enlaces:
                        href = enlace.get_attribute("href")
                        
                        if href and "modalidad-" in href:
                            # Extraer el patrón usando regex libreria para trabajar con expresiones regulares
                            match = re.search(r"modalidad-(\w+)", href)
                            if match:
                                modalidad_texto = match.group(1).lower() # grupo 1 es el texto que queremos, y lo convertimos a minusculas para evitar problemas de mayusculas o minusculas
                                
                                # Convertir el texto encontrado a una modalidad válida
                                if modalidad_texto == "remoto":
                                    modalidad_encontrada = "Remoto"
                                elif modalidad_texto == "presencial":
                                    modalidad_encontrada = "Presencial"
                                elif modalidad_texto == "hibrido": #Este es especial ## en hrf esta en mnusculas y sin tilde
                                    modalidad_encontrada = "Híbrido"

                                
                                break
                
                if modalidad_encontrada:
                    break
            
            if not modalidad_encontrada:
                print("No se encontró la modalidad en los enlaces de navegación")
                
        except Exception as e:
            print(f"Error al buscar la modalidad: {e}")
        
        ## Fin del try de navegacion

        resultados.extend([modalidad_encontrada])
        
    except Exception as e:
        print(f"Error general en la página {url}: {e}")
        

driver.quit()

# Mostrar resultados finales

print(f'Total de modalidades encontradas: {len(resultados)}')
```

## Crear el .CSV


```python
df1= pd.DataFrame(DATA, columns=["Job_title", "Description", "Distric"])
df2= pd.DataFrame(resultados, columns=["Modalidad"])
df3= pd.concat([df1, df2], axis=1) #Aquí utilizo Axis=1 para concatenar las columnas, si pones 0 (fila) 1 (columnas)
df3.to_csv("000246574_hw1_2025_1.csv", index=False, encoding="utf-8-sig")  
```
