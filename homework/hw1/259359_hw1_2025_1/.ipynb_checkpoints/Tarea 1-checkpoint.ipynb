```python
#importo las librerías que se instalaron arriba
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
import re
import pandas as pd
```

```python
##ejecuto el driver y abro la pagina de bumeran
driver = webdriver.Chrome()
url = "https://www.bumeran.com.pe/empleos.html"
driver.get(url)
driver.maximize_window()
time.sleep(5) #esperamos a que se abra la pestaña
driver.execute_script("document.body.style.zoom='100%'") #ampliamos la ventana
```

----------FILTROS------------------------

```python
# Abriendo el filtro de fecha de publicación y seleccionando los últimos 15 días
driver.find_element(By.NAME, 'icon-light-calendar').click()
time.sleep(3)

fecha_15_dias = driver.find_element(By.XPATH, "//button[contains(text(), 'Menor a 15 días')]")
fecha_15_dias.click()
time.sleep(3)

# Entrando al menú de áreas laborales y eligiendo "Tecnología, Sistemas y Telecomunicaciones"
driver.find_element(By.NAME, 'icon-light-cube').click()
time.sleep(3)

area_tecnologia = driver.find_element(By.XPATH, "//button[contains(text(), 'Tecnología, Sistemas y Telecomunicaciones')]")
area_tecnologia.click()
time.sleep(3)

# Accediendo al menú de subáreas y seleccionando "Programación"
driver.find_element(By.NAME, 'icon-light-layers').click()
time.sleep(3)

subarea_programacion = driver.find_element(By.XPATH, "//button[contains(text(), 'Programación')]")
subarea_programacion.click()
time.sleep(7)

# Configurando el filtro de ubicación en "Lima"
driver.find_element(By.XPATH, "//button//i[@name='icon-light-location-pin']/..").click()
time.sleep(9)

lima = driver.find_element(By.XPATH, "//button[contains(text(), 'Lima')]")
lima.click()
time.sleep(10)

# Ajustando la preferencia de jornada laboral a "Full-time"
driver.find_element(By.NAME, 'icon-light-bookmark').click()
time.sleep(10)

full_time = driver.find_element(By.XPATH, "//button[contains(text(), 'Full-time')]")
full_time.click()
```

```python
# Esperar a que los botones con enlaces de empleo carguen en la página
WebDriverWait(driver, 8).until(EC.presence_of_all_elements_located((By.CLASS_NAME, "sc-zDqdV")))
# Buscar todos los botones que contienen los enlaces de las ofertas
botones_enlaces = driver.find_elements(By.CLASS_NAME, "sc-zDqdV")
# Extraer los enlaces de cada botón y guardarlos en una lista
enlaces_empleos = [boton.get_attribute("href") for boton in botones_enlaces]
```

```python
enlaces_empleos # comprobamos
```

```python
#✅ Stage 2: Scrape Job Details
# Iniciamos el navegador y lo maximizamos
driver = webdriver.Chrome()
driver.maximize_window()

# Lista para almacenar la información
datos_empleos = []

# Función para obtener el distrito
def obtener_distrito(driver):
    xpaths_distrito = [
        "//div[contains(@class, 'ubicacion')]//h2",
        "//li[contains(@class, 'ubicacion')]//a"
    ]
    for xpath in xpaths_distrito:
        elementos = driver.find_elements(By.XPATH, xpath)
        if elementos:
            return elementos[0].text.split(",")[0].strip()  # Extraemos solo el distrito
    return "Distrito no encontrado"

# Función para obtener la modalidad de trabajo
def obtener_modalidad(driver):
    xpaths_modalidad = [
        "//ul[contains(@class, 'modalidad')]//li/a",
        "//div[contains(@class, 'detalles')]//ul/li"
    ]
    for xpath in xpaths_modalidad:
        elementos = driver.find_elements(By.XPATH, xpath)
        if elementos:
            return elementos[0].text.strip()
    return "No especificado"

# Función para obtener la descripción del empleo
def obtener_descripcion(driver):
    try:
        descripcion_completa = driver.find_element(By.ID, "ficha-detalle").text
        descripcion, _, _ = descripcion_completa.partition("Beneficios")  # Toma solo lo anterior a "Beneficios"
        return descripcion.strip()
    except:
        return "Descripción no disponible"

# Recorremos los enlaces y extraemos la información
for enlace in lista_enlaces:
    try:
        driver.get(enlace)

        # Extraer título
        titulo = WebDriverWait(driver, 8).until(
            EC.presence_of_element_located((By.TAG_NAME, 'h1'))
        ).text

        # Extraer información usando funciones
        distrito = obtener_distrito(driver)
        modalidad = obtener_modalidad(driver)
        descripcion = obtener_descripcion(driver)

        # Guardamos los datos
        datos_empleos.append([titulo, descripcion, distrito, modalidad, enlace])

        print(f"Datos obtenidos de: {enlace}")

    except Exception as e:
        print(f"Error al procesar {enlace}: {e}")

# Cerrar navegador
driver.quit()

# Guardar en un archivo Excel
df = pd.DataFrame(datos_empleos, columns=["Título", "Descripción", "Distrito", "Modalidad", "Enlace"])
df.to_csv("Trabajos_Bumeran.csv", index=False, encoding="utf-8", sep="|")

print("Extracción completada. Datos guardados en 'Trabajos_Bumeran.xlsx'.")
```
