```python

# Esperamos a que carguen los elementos con la clase específica
WebDriverWait(driver, 10).until(
    EC.presence_of_element_located((By.CLASS_NAME, "sc-eCXBzT.cxJokR"))
)

# Lista para guardar los enlaces
job_links = []

# Encontrar todos los recuadros de trabajo
job_boxes = driver.find_elements(By.CLASS_NAME, "sc-eCXBzT.cxJokR")
print(len(job_boxes)) # con esto obtengo cuantas cajitas hay, a partir de ello jalaré los enlaces


```

```python
# Definir espera explícita
wait = WebDriverWait(driver, 10)

# Lista para almacenar los enlaces de los empleos
job_links = []

# Iterar sobre los primeros 20 elementos
for i in range(2, 22):  # Desde 2 hasta 20, porque inicia en el link 2
    try:
        xpath = f'//*[@id="listado-avisos"]/div[{i}]/a'
        
        # Esperar a que el enlace esté presente y obtenerlo
        link_element = wait.until(EC.presence_of_element_located((By.XPATH, xpath)))
        
        job_link = link_element.get_attribute("href")
        if job_link:
            job_links.append(job_link)
    except Exception as e:
        print(f"No se encontró enlace en el recuadro {i}: {e}")

# Imprimir los enlaces encontrados
for link in job_links:
    print(link)

print (len(job_links))
```

```python

descripciones=[]
nombres=[]
distritos=[]
modalidades=[]
for link in job_links:
    driver = webdriver.Chrome()
    driver.get(link)
    driver.maximize_window()
    ######################################################################################################
    # NOMBRE DEL TRABAJO
    try:
        # Intentar con CSS_SELECTOR
        job_name_text = driver.find_element(By.CSS_SELECTOR, "#header-component > div.sc-giPFmd.eAXioh > div > div.sc-doIlLK.iMoClT > h1").text
    except Exception:
        # Si falla, intentar con XPATH
        xpath_1 = '//*[@id="header-component"]/div[1]/div/div[1]/h1'
        job_name_text = WebDriverWait(driver, 30).until(EC.presence_of_element_located((By.XPATH, xpath_1))).text

    nombres.append(job_name_text)
    ######################################################################################################
    # DISTRITO

    try:
        # Intentar con CSS_SELECTOR
        district_name_text = driver.find_element(By.CSS_SELECTOR, "#ficha-detalle > div:nth-child(2) > div > div.sc-gWIwhj.zkMuK > div:nth-child(1) > div:nth-child(2) > div > div > li > a > h2").text
    except Exception:
        # Si falla, intentar con XPATH
        xpath_2 = '//*[@id="ficha-detalle"]/div[2]/div/div[1]/div[1]/div[2]/div/div/li/a/h2'
        district_name_text = WebDriverWait(driver, 30).until(EC.presence_of_element_located((By.XPATH, xpath_2))).text

    distritos.append(district_name_text)

    ######################################################################################################
    # MODALIDAD
    try:
        # Intentar con CSS_SELECTOR
        modalidad_name_text = driver.find_element(By.CSS_SELECTOR, "#ficha-detalle > div:nth-child(2) > div > div.sc-gWIwhj.zkMuK > div:nth-child(7) > div > ul > div:nth-child(1) > li:nth-child(1) > a > p").text
    except Exception:
        try:
            # Si falla, intentar con XPATH
            xpath_3 = '//*[@id="ficha-detalle"]/div[2]/div/div[1]/div[4]/div/ul/div[1]/li[1]/a/p'
            modalidad_name_text = WebDriverWait(driver, 30).until(EC.presence_of_element_located((By.XPATH, xpath_3))).text
        except:
            xpath_3_1 = '//*[@id="ficha-detalle"]/div[2]/div/div[1]/div[5]/div/ul/div[1]/li[1]/a/p'
            modalidad_name_text = WebDriverWait(driver, 30).until(EC.presence_of_element_located((By.XPATH, xpath_3_1))).text
    
        

    modalidades.append(modalidad_name_text)
    ######################################################################################################
    # DESCRIPCION DEL TRABAJO
    try:
        # En ficha detalle se encuentra todo lo que queremos sacar
        detalle = WebDriverWait(driver, 10).until(
            EC.presence_of_element_located((By.ID, "ficha-detalle"))
        )
        
        # Extraemos el html del objeto pues este contiene todos los elementos
        # no puedo convertirlo en texto y de ahi partirlo porque se borra el elemento de rayita que tiene 
        detalle_html = detalle.get_attribute("outerHTML")
        
        # el separador será la rayita que existe entre la descripción del trabajo y las características de este

        separador_element = WebDriverWait(driver, 10).until(
            EC.presence_of_element_located((By.CSS_SELECTOR, ".sc-clieFk.jjxmCV"))
        )
        
        separador_html = separador_element.get_attribute("outerHTML")  # Obtener HTML del separador
        
        # Cortar el HTML hasta el separador
        end_index = detalle_html.find(separador_html)
        
        if end_index != -1:
            html_recortado = detalle_html[:end_index]  # Tomar solo la parte antes del separador
            
            # importamos la sigueinte liberia para convertir html a texto
            from html.parser import HTMLParser
            
            class MLStripper(HTMLParser):
                def __init__(self):
                    super().__init__()
                    self.text = []
                def handle_data(self, data):
                    self.text.append(data)
                def get_data(self):
                    return ' '.join(self.text)

            stripper = MLStripper()
            stripper.feed(html_recortado)
            texto_extraido = stripper.get_data()
            
            #print(texto_extraido.strip())  # Mostrar solo el texto antes del separador
            descripciones.append(texto_extraido.strip())
        else:
            print("No se encontró el separador en el HTML.")

    except Exception as e:
        print("Error:", e)

    finally:
        driver.quit()  # Cerrar el navegador
    




print('LINKS LEIDOS CORRECTAMENTE '+str(len(job_links)))





#//*[@id="header-component"]/div[1]/div/div[1]/h1
```
