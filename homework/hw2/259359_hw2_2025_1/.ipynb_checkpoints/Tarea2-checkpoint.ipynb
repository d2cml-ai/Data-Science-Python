```python
# --------------------- Extracción de Datos de Codeforces -----------------------
import requests
import pandas as pd
from datetime import datetime
import time

# --- Parámetros base ---
USER_HANDLES = ["tourist", "benq", "Radewoosh"]  # Handles a analizar
START_DATE = datetime(2024, 7, 1).timestamp()
END_DATE = datetime(2024, 12, 31).timestamp()
KEYWORDS = ["Hello", "Round", "Good Bye"]

# --- Paso 1: Obtener lista de concursos válidos ---
def get_filtered_contests():
    url = "https://codeforces.com/api/contest.list"
    res = requests.get(url).json()
    if res["status"] != "OK":
        raise Exception("Error al obtener la lista de concursos.")
    
    contests = res["result"]
    filtered = [
        c for c in contests
        if c["phase"] == "FINISHED"
        and START_DATE <= c["startTimeSeconds"] <= END_DATE
        and any(keyword in c["name"] for keyword in KEYWORDS)
    ]
    return filtered

# --- Paso 2: Obtener información de cada concurso con standings y problemas ---
def get_contest_details(contest_id):
    url = f"https://codeforces.com/api/contest.standings?contestId={contest_id}&from=1&count=1"
    res = requests.get(url).json()
    if res["status"] != "OK":
        print(f"⚠️ Falló standings para contest {contest_id}")
        return None
    return res["result"]["contest"], res["result"]["problems"]

# --- Paso 3: Obtener lista de usuarios que participaron (solo rated) ---
def get_rated_users():
    url = "https://codeforces.com/api/user.ratedList?activeOnly=false"
    res = requests.get(url).json()
    if res["status"] != "OK":
        raise Exception("Error al obtener rated users.")
    return res["result"]

# --- Paso 4: Obtener envíos de un usuario en un concurso específico ---
def get_user_submissions(contest_id, handle):
    url = f"https://codeforces.com/api/contest.status?contestId={contest_id}&handle={handle}"
    res = requests.get(url).json()
    return res["result"] if res["status"] == "OK" else []

# --- Paso 5: Obtener cambios de rating por concurso ---
def get_rating_changes(contest_id):
    url = f"https://codeforces.com/api/contest.ratingChanges?contestId={contest_id}"
    res = requests.get(url).json()
    return res["result"] if res["status"] == "OK" else []

# --- Ejecutar todo lo anterior ---
contests = get_filtered_contests()
print(f"\n🧠 Concursos válidos encontrados: {len(contests)}")

all_data = []

for contest in contests:
    cid = contest["id"]
    cname = contest["name"]
    ctime = datetime.fromtimestamp(contest["startTimeSeconds"]).strftime('%Y-%m-%d %H:%M:%S')
    
    print(f"\n🔍 Procesando: {cid} - {cname}")
    contest_info, problems = get_contest_details(cid)
    rating_changes = get_rating_changes(cid)
    rating_map = {r['handle']: r for r in rating_changes}
    
    for handle in USER_HANDLES:
        submissions = get_user_submissions(cid, handle)
        time.sleep(0.5)  # Respetar la API
        
        entry = {
            "handle": handle,
            "contest_id": cid,
            "contest_name": cname,
            "start_time": ctime,
            "rating_before": rating_map.get(handle, {}).get("oldRating", None),
            "rating_after": rating_map.get(handle, {}).get("newRating", None),
        }

        # Datos por problema (A, B, C...)
        problem_letters = ["A", "B", "C"]
        accepted = sorted([s for s in submissions if s.get("verdict") == "OK"], key=lambda s: s["relativeTimeSeconds"])
        prev_time = 0

        for p in problem_letters:
            prob_subs = [s for s in accepted if s['problem']['index'] == p]
            if prob_subs:
                first = prob_subs[0]
                rel_time = first['relativeTimeSeconds']
                entry[f"finished_{p}"] = True
                entry[f"relative_time_{p}"] = rel_time
                entry[f"time_to_answer_{p}"] = rel_time - prev_time if prev_time else rel_time
                entry[f"{p}_language"] = first.get('programmingLanguage', 'Unknown')
                prev_time = rel_time
            else:
                entry[f"finished_{p}"] = False
                entry[f"relative_time_{p}"] = None
                entry[f"time_to_answer_{p}"] = None
                entry[f"{p}_language"] = None

        all_data.append(entry)

# --- Convertir a DataFrame ---
df = pd.DataFrame(all_data)
print("\n✅ Data extraída exitosamente. Muestra:")
print(df.head())

#Solo considera concursos válidos entre julio y diciembre de 2024.

#Solo incluye concursos que contienen "Hello", "Round" o "Good Bye".

#Extrae:

#Info del concurso y problemas (contest.standings)

#Cambios de rating (contest.ratingChanges)

#Submissions por usuario (contest.status)

#Procesa problemas por letra (A, B, C) con relative_time y time_to_answer.
```

```python
# ------------------- Data Wrangling: Recolección, limpieza y transformación -------------------
import pandas as pd

# Asumimos que el DataFrame original ya está en df

# ------------------ LIMPIEZA BÁSICA ------------------
# Asegurarnos de que columnas clave estén bien tipadas
df['contest_start_time'] = pd.to_datetime(df['contest_start_time'])  # Convertir timestamps a datetime

# Convertir ratings a numéricos (por si hay valores None o vacíos)
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')
df['max_rating'] = pd.to_numeric(df['max_rating'], errors='coerce')
df['rating_achieved'] = pd.to_numeric(df['rating_achieved'], errors='coerce')

# ------------------ VARIABLES DERIVADAS POR PROBLEMA ------------------
problem_letters = ['A', 'B', 'C']

# Aseguramos que cada usuario y concurso tenga un cálculo correcto
for handle in df['author_handle'].unique():
    for cid in df['contest_id'].unique():
        subset = df[(df['author_handle'] == handle) & (df['contest_id'] == cid)]

        if len(subset) == 1:
            idx = subset.index[0]
            prev_time = 0

            for p in problem_letters:
                rt_col = f'relative_time_{p}'
                tt_col = f'time_to_answer_{p}'

                rel_time = df.at[idx, rt_col]

                if pd.notna(rel_time):
                    if prev_time == 0:
                        df.at[idx, tt_col] = rel_time
                    else:
                        df.at[idx, tt_col] = rel_time - prev_time
                    prev_time = rel_time
                else:
                    df.at[idx, tt_col] = None

# Convertimos finished_n a booleano
for p in problem_letters:
    df[f'finished_{p}'] = df[f'finished_{p}'].astype(bool)

# ------------------ RESULTADO ------------------
print("✅ Wrangling completado. Muestra:")
print(df.head())
```

```python
#Visualizaciones y Análisis
import matplotlib.pyplot as plt
import seaborn as sns

# --- Histograma del rating logrado ---
plt.figure(figsize=(8, 5))
sns.histplot(df['rating_achieved'].dropna(), kde=True, bins=10, color='skyblue')
plt.title("Distribución del Rating Alcanzado")
plt.xlabel("Rating")
plt.ylabel("Frecuencia")
plt.show()

# --- Boxplot: Lenguaje más usado vs. tiempo de respuesta (problema A) ---
plt.figure(figsize=(10, 6))
sns.boxplot(x='A_language', y='time_to_answer_A', data=df)
plt.xticks(rotation=45)
plt.title("Tiempo para Resolver Problema A según Lenguaje")
plt.show()

# --- Scatterplot: Rating vs Tiempo para resolver problema B ---
plt.figure(figsize=(8, 5))
sns.scatterplot(x='rating', y='time_to_answer_B', data=df)
sns.regplot(x='rating', y='time_to_answer_B', data=df, scatter=False, color='red')
plt.title("Rating vs. Tiempo para Resolver Problema B")
plt.xlabel("Rating")
plt.ylabel("Tiempo para Responder (s)")
plt.show()
```
