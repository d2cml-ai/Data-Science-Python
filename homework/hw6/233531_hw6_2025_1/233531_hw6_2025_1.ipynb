```python
from tensorflow.keras.datasets import fashion_mnist

# Cargar datos de entrenamiento y prueba
(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()
```

```python
# Normalización
x_train = x_train / 255.0
x_test = x_test / 255.0

# Aplanar imágenes
x_train_flat = x_train.reshape(-1, 28*28)
x_test_flat = x_test.reshape(-1, 28*28)
```

```python
from tensorflow.keras import Sequential, Input
from tensorflow.keras.layers import Dense

model = Sequential([
    Input(shape=(784,)),  # Declarar explícitamente la entrada
    Dense(128, activation='relu'),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')
])
```

```python
import time

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

start_time_cpu = time.time()

model.fit(x_train_flat, y_train, epochs=5, validation_split=0.1)

end_time_cpu = time.time()
cpu_time = end_time_cpu - start_time_cpu
print(f"⏱ Tiempo de entrenamiento con CPU: {cpu_time:.2f} segundos")
```

```python
with open("resultados_cpu.txt", "w") as f:
    f.write(f"{cpu_time}")
```

Cambio a GPU

```python
from tensorflow.keras.datasets import fashion_mnist

# Cargar datos de entrenamiento y prueba
(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()

# Normalización
x_train = x_train / 255.0
x_test = x_test / 255.0

# Aplanar imágenes
x_train_flat = x_train.reshape(-1, 28*28)
x_test_flat = x_test.reshape(-1, 28*28)
```

```python
# Redefinir el modelo para usarlo limpio en GPU
from tensorflow.keras import Sequential, Input
from tensorflow.keras.layers import Dense

model_gpu = Sequential([
    Dense(128, activation='relu', input_shape=(784,)),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')
])

import time

model_gpu.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])

start_time_gpu = time.time()

model_gpu.fit(x_train_flat, y_train, epochs=5, validation_split=0.1)

end_time_gpu = time.time()
gpu_time = end_time_gpu - start_time_gpu
print(f"⚡ Tiempo de entrenamiento con GPU: {gpu_time:.2f} segundos")
```

```python
cpu_time = 35.27

print(f"CPU: {cpu_time:.2f} segundos")
print(f"GPU: {gpu_time:.2f} segundos")

diff = cpu_time - gpu_time
print(f"Diferencia: {diff:.2f} segundos a favor de {'GPU' if diff > 0 else 'CPU'}")
```

```python
with open("resultados_tiempos.txt", "w") as f:
    f.write(f"Tiempo CPU: {cpu_time:.2f} segundos\n")
    f.write(f"Tiempo GPU: {gpu_time:.2f} segundos\n")
    f.write(f"Diferencia: {diff:.2f} segundos a favor de {'GPU' if diff > 0 else 'CPU'}")
```

```python
# ✅ ENTRENAMIENTO TOXIC COMMENTS CON BERT-TINY (OPTIMIZADO PARA CPU/GPU EN COLAB)

# === 1. Cargar CSV original descargado de Kaggle ===
import pandas as pd

csv_path = "/content/train.csv"  # Asegúrate de subir este archivo a Colab
df = pd.read_csv(csv_path)

# === 2. Crear la variable binaria `label` (1 si toxic >= 0.5, sino 0) ===
df["label"] = df["toxic"].apply(lambda x: int(x >= 0.5))

# === 3. Filtrar columnas necesarias ===
df = df[["comment_text", "label"]].dropna()
```

```python
# === 4. Tokenizar los textos con `bert-tiny` ===
!pip install -q transformers
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("prajjwal1/bert-tiny")

encodings = tokenizer(
    list(df["comment_text"]),
    truncation=True,
    padding=True,
    max_length=32,  # más reducido aún para mayor velocidad
    return_tensors="pt"
)

# === 5. Preparar tensores de entrada y etiquetas ===
import torch
from torch.utils.data import TensorDataset, DataLoader, random_split

labels = torch.tensor(df["label"].values)

dataset = TensorDataset(
    encodings["input_ids"],
    encodings["attention_mask"],
    labels
)

# === 6. Usar subconjunto reducido para comparación CPU vs GPU ===
subset_size = 400
train_size = 300
val_size = 100

subset, _ = random_split(dataset, [subset_size, len(dataset) - subset_size])
train_dataset, val_dataset = random_split(subset, [train_size, val_size])

train_dataloader = DataLoader(train_dataset, batch_size=16, shuffle=True)
val_dataloader = DataLoader(val_dataset, batch_size=16)
```

```python
# === 7. Cargar modelo BERT-TINY para clasificación binaria ===
from transformers import AutoModelForSequenceClassification

model = AutoModelForSequenceClassification.from_pretrained("prajjwal1/bert-tiny", num_labels=2)

# === 8. Configurar dispositivo (CPU o GPU) ===
import time

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)

# === 9. Configurar optimizador ===
optimizer = torch.optim.AdamW(model.parameters(), lr=5e-5)
```

```python
# === 10. Entrenamiento ===
epochs = 2
model.train()
start_time = time.time()

for epoch in range(epochs):
    total_loss = 0
    print(f"\n🔁 Epoch {epoch+1}")
    for batch in train_dataloader:
        input_ids, attention_mask, labels = [b.to(device) for b in batch]

        optimizer.zero_grad()
        outputs = model(input_ids=input_ids, attention_mask=attention_mask, labels=labels)
        loss = outputs.loss
        total_loss += loss.item()
        loss.backward()
        optimizer.step()

    print(f"Loss promedio de entrenamiento: {total_loss / len(train_dataloader):.4f}")

end_time = time.time()
print(f"\n⏱ Tiempo total: {end_time - start_time:.2f} segundos usando {'GPU' if torch.cuda.is_available() else 'CPU'}")

# === 11. Evaluación ===
from sklearn.metrics import accuracy_score, f1_score

model.eval()
preds, true_labels = [], []

with torch.no_grad():
    for batch in val_dataloader:
        input_ids, attention_mask, labels = [b.to(device) for b in batch]
        outputs = model(input_ids=input_ids, attention_mask=attention_mask)
        logits = outputs.logits
        predictions = torch.argmax(logits, dim=-1)

        preds.extend(predictions.cpu().numpy())
        true_labels.extend(labels.cpu().numpy())

accuracy = accuracy_score(true_labels, preds)
f1 = f1_score(true_labels, preds)

print(f"\n✅ Accuracy: {accuracy:.4f}")
print(f"✅ F1 Score: {f1:.4f}")
```
