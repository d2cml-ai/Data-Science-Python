### Entrenar una red neuronal convolucional simple (CNN) para clasificar imágenes de ropa, y comparar el tiempo de entrenamiento en CPU vs GPU

## # ⏱ Importar librerías y configurar entorno

```python

import time
import tensorflow as tf
from tensorflow.keras.datasets import fashion_mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
import matplotlib.pyplot as plt
```

### Usamos el dataset Fashion MNIST que viene incorporado en Keras. Contiene imágenes de 28x28 pixeles de distintos tipos de ropa.



## 🔢 Cargar dataset Y 🧼 Normalizar y adaptar forma

```python

(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()


x_train = x_train.reshape(-1, 28, 28, 1).astype("float32") / 255.0
x_test = x_test.reshape(-1, 28, 28, 1).astype("float32") / 255.0

```

### Luego normalizamos los valores de los pixeles dividiéndolos entre 255, y le damos forma (28, 28, 1) para que sean compatibles con una CNN.

## 🧠 Definir modelo simple CNN

### Creamos una CNN sencilla que toma las imágenes como entrada, extrae características, y finalmente clasifica en 10 clases. Usamos relu y softmax.

```python

def create_model():
    model = Sequential([
        Conv2D(32, (3,3), activation='relu', input_shape=(28,28,1)),
        MaxPooling2D(2,2),
        Flatten(),
        Dense(64, activation='relu'),
        Dense(10, activation='softmax')
    ])
    model.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])
    return model

```

## 🖥️ Entrenar en CPU

```python

with tf.device('/CPU:0'):
    model_cpu = create_model()
    start_cpu = time.time()
    model_cpu.fit(x_train, y_train, epochs=5, validation_split=0.1, verbose=1)
    cpu_time = time.time() - start_cpu

```

## ⚡ Entrenar en GPU

```python

with tf.device('/GPU:0'):
    model_gpu = create_model()
    start_gpu = time.time()
    model_gpu.fit(x_train, y_train, epochs=5, validation_split=0.1, verbose=1)
    gpu_time = time.time() - start_gpu
```


## 📊 Mostrar tiempos

```python

print(f"Tiempo en CPU: {cpu_time:.2f} segundos")
print(f"Tiempo en GPU: {gpu_time:.2f} segundos")
```

## 📚 Importar librerías

```python

import torch
import time
from datasets import load_dataset
from transformers import DistilBertTokenizerFast, DistilBertForSequenceClassification, Trainer, TrainingArguments
```

### Originalmente, íbamos a usar dataset jigsaw_toxicity_pred, pero ese dataset requiere descarga manual desde Kaggle. Para simplificar y evitar bloqueos, usamos un dataset público alternativo llamado civil_comments, que tiene la misma estructura (columna text y toxicity), pero está disponible directamente desde Hugging Face.

```python
# 🔄 Cargar dataset (usamos una muestra)
dataset = load_dataset("jigsaw_toxicity_pred", split='train[:1000]')

# 🔁 Convertir a variable binaria
def preprocess(example):
    example['label'] = int(example['toxicity'] >= 0.5)
    return example

dataset = dataset.map(preprocess)

# 🧼 Tokenización
tokenizer = DistilBertTokenizerFast.from_pretrained('distilbert-base-uncased')

def tokenize(example):
    return tokenizer(example['text'], padding='max_length', truncation=True, max_length=128)

tokenized_dataset = dataset.map(tokenize, batched=True)
tokenized_dataset.set_format(type='torch', columns=['input_ids', 'attention_mask', 'label'])
```

## ✅ Cargar el dataset público (solo 1000 ejemplos), convertir toxicidad a variable binaria, y tokenizar con DistilBERT

### Convertimos la columna "toxicity" en una variable binaria "label":

### 1 si el valor es mayor o igual a 0.5

### 0 si es menor

### Simplpificando la tarea a una clasificación binaria.

### Y usamos el tokenizador de DistilBERT (distilbert-base-uncased), que transforma los textos en vectores que el modelo puede entender.

```python
# ✅ 1. Cargar el dataset público
dataset = load_dataset("civil_comments", split="train[:1000]")

# ✅ 2. Convertir toxicidad a variable binaria
def preprocess(example):
    example["label"] = int(example["toxicity"] >= 0.5)
    return example

dataset = dataset.map(preprocess)

# ✅ 3. Tokenización con DistilBERT
tokenizer = DistilBertTokenizerFast.from_pretrained('distilbert-base-uncased')

def tokenize(example):
    return tokenizer(
        example['text'],
        padding='max_length',
        truncation=True,
        max_length=128
    )

tokenized_dataset = dataset.map(tokenize, batched=True)
tokenized_dataset.set_format(type='torch', columns=['input_ids', 'attention_mask', 'label'])
```

## 🧠 Creación del modelo y entrenamiento en CPU

```python
# 🚫 Desactivar W&B para evitar que pida login
import os
os.environ["WANDB_DISABLED"] = "true"
# 🧠 Modelo
model = DistilBertForSequenceClassification.from_pretrained('distilbert-base-uncased', num_labels=2)

# 🖥️ Entrenamiento en CPU
training_args_cpu = TrainingArguments(
    output_dir="./results_cpu",
    num_train_epochs=2,
    per_device_train_batch_size=8,
    logging_steps=10,
    no_cuda=True,
    logging_dir="./logs_cpu",
)

trainer_cpu = Trainer(
    model=model,
    args=training_args_cpu,
    train_dataset=tokenized_dataset
)

start_cpu = time.time()
trainer_cpu.train()
cpu_time = time.time() - start_cpu
print(f"Tiempo en CPU (BERT): {cpu_time:.2f} segundos")


```

## ⚡ Entrenamiento en GPU (nuevo modelo para reiniciar pesos)

```python
# 🚫 Desactivar W&B
import os
os.environ["WANDB_DISABLED"] = "true"

# ⚡ Entrenamiento en GPU (nuevo modelo para reiniciar pesos)
model_gpu = DistilBertForSequenceClassification.from_pretrained('distilbert-base-uncased', num_labels=2)

training_args_gpu = TrainingArguments(
    output_dir="./results_gpu",
    num_train_epochs=2,
    per_device_train_batch_size=8,
    logging_steps=10,
    logging_dir="./logs_gpu",
)

trainer_gpu = Trainer(
    model=model_gpu,
    args=training_args_gpu,
    train_dataset=tokenized_dataset
)

start_gpu = time.time()
trainer_gpu.train()
gpu_time = time.time() - start_gpu

print(f"Tiempo en GPU (BERT): {gpu_time:.2f} segundos")
```
