```python
import tensorflow as tf
from tensorflow.keras import layers, models
import numpy as np
import time

# 1. Cargar Fashion-MNIST dataset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()

# 2. Preprocesar datos
# Normalización de las imágenes
x_train, x_test = x_train / 255.0, x_test / 255.0

# Aplanar las imágenes (28x28 píxeles a 784 dimensiones)
x_train = x_train.reshape(-1, 28*28)
x_test = x_test.reshape(-1, 28*28)

# Verificar las formas de los datos
print(f"x_train shape: {x_train.shape}")  # (60000, 784)
print(f"x_test shape: {x_test.shape}")    # (10000, 784)

# 3. Diseñar un modelo de red neuronal simple
model = models.Sequential([
    layers.Dense(128, activation='relu', input_shape=(28*28,)),
    layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# 4. Entrenamiento en CPU (sin especificar explícitamente el dispositivo)
print("Entrenando en CPU...")
start_time = time.time()
model.fit(x_train, y_train, epochs=5, batch_size=64)
cpu_time = time.time() - start_time
print(f"Tiempo de entrenamiento en CPU: {cpu_time:.2f} segundos")

# 5. Entrenamiento en GPU (TensorFlow seleccionará automáticamente la GPU si está disponible)
print("Entrenando en GPU...")
start_time = time.time()
model.fit(x_train, y_train, epochs=5, batch_size=64)
gpu_time = time.time() - start_time
print(f"Tiempo de entrenamiento en GPU: {gpu_time:.2f} segundos")

# 6. Comparar resultados
print(f"Mejora de velocidad: {cpu_time / gpu_time:.2f}x más rápido en GPU")

```
