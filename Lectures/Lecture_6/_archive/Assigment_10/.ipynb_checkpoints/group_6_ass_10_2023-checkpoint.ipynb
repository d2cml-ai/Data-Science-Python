### 1.  Generate a map of the total institutes' vacancies by the department.

```python
# Crear GeoDataFrame. Definir sistema de coordenadas y crear puntos a partir de las coordenadas (lon y lat)
institutos_geo = gpd.GeoDataFrame( institutos, crs = "EPSG: 4326", 
                                             geometry = gpd.points_from_xy( institutos.nlong_ie, institutos.nlat_ie))

# Importar archivo shp de departamentos y unirlo con la data de institutos_geo
shp_dpt = gpd.read_file( r'../../_data/INEI_LIMITE_DEPARTAMENTAL/INEI_LIMITE_DEPARTAMENTAL.shp')
intersect_inst_geo = gpd.overlay( institutos_geo, shp_dpt, how = "intersection")

# Definir lista de variables: meta de admision, total postulantes, total ingresantes, total matriculados
vars_sum = [ "ltimoden_metaatencion", "cuentadeid_postulante_procesoadm", "sumaden_flagingresante", "sumaden_flagmatriculado"]

# Agrupar los datos según departamento (CCDD) y calcular la cantidad total de casos en las variables por cada departamento 
tot_dpt = intersect_inst_geo.groupby( ["CCDD"], as_index = False)[ vars_sum ].sum()
tot_dpt_shp = shp_dpt.merge( tot_dpt, on = "CCDD")
tot_dpt_shp
```

```python
# government palace coordinates

lat_palacio  = -10.0757538
long_palacio = -76.9863174
zoom_start   = 5.4

z = fm.Map(location = [lat_palacio, long_palacio], tiles = 'cartodbpositron', zoom_start = zoom_start)

fm.Choropleth(
    geo_data=tot_dpt_shp,
    data=tot_dpt_shp,
    columns=['CCDD', 'ltimoden_metaatencion'],
    key_on="feature.properties.CCDD",
    fill_color="YlOrRd",
    fill_opacity=0.8,
    line_opacity=0.2,
    legend_name="Total Vacancies",
    smooth_factor=0,
    Highlight= True,
    line_color = "#0000",
    overlay=True,
    nan_fill_color = "White"  # fill white missing values 
    ).add_to(z)

z
```

### 2. Interactive map of markes  show total vacancies, applicants, entrants, and enrolled information 

```python
# analizar estructura del df que contiene los casos por cada instituto
institutos_geo.head(3)
```

```python
# crear una función para establecer el estilo visual de las tablas (pop-up)
def visual_html(i):
    
    # informacion por cada instituto
    vacancies  = institutos_geo['ltimoden_metaatencion'].iloc[i]                             
    applicants = institutos_geo['cuentadeid_postulante_procesoadm'].iloc[i]                           
    entrants   = institutos_geo['sumaden_flagingresante'].iloc[i]  
    enrolled   = institutos_geo['sumaden_flagmatriculado'].iloc[i]  
    
    # definir los colores para cada columna de la tabla
    left_col_colour = "#1DA1F2"
    right_col_colour = "#EFEFEF"
    
    html = """<!DOCTYPE html>
<html>

<head>
    <p><b> Información del instituto </b></p>

</head>
    <table style="height: 126px; width: 350px;">  <!-- Comment: Create a teable. -->

<!-- Add information  -->

<tbody> 
<tr>

<!-- Add color by column -->

<td style="background-color: """+ left_col_colour +"""; text-align: center;"><span style="color: #ffffff;">Total vacantes</span></td>
<td style="width: 150px;background-color: """+ right_col_colour +"""; text-align: center;">{}</td>""".format(vacancies) + """
</tr>
<tr>
<td style="background-color: """+ left_col_colour +"""; text-align: center;"><span style="color: #ffffff;">Total postulantes</span></td>
<td style="width: 150px;background-color: """+ right_col_colour +"""; text-align: center;">{}</td>""".format(applicants) + """
</tr>
<tr>
<td style="background-color: """+ left_col_colour +"""; text-align: center;"><span style="color: #ffffff;">Total ingresantes</span></td>
<td style="width: 150px;background-color: """+ right_col_colour +"""; text-align: center;">{}</td>""".format(entrants) + """
</tr>
<tr>
<td style="background-color: """+ left_col_colour +"""; text-align: center;"><span style="color: #ffffff;">Total matriculados</span></td>
<td style="width: 150px;background-color: """+ right_col_colour +"""; text-align: center;">{}</td>""".format(enrolled) + """
</tr>
<tr>


</tbody>
</table>
</html>
"""
    return html
```

```python
# crear mapa
m = fm.Map(location = [-10.0757538, -76.9863174], zoom_start = 5.5)

# iterar por cada instituto (fila) del dataframe
for i in range(0,len(institutos_geo)):
    
    html = visual_html(i)
    
    iframe = br.element.IFrame(html=html,width=350,height=200)
    popup = fm.Popup(iframe, parse_html=True)
    
    # definir la posición y el estilo de los markers
    fm.Marker([institutos_geo['nlat_ie'].iloc[i], institutos_geo['nlong_ie'].iloc[i]],
                  popup=popup, icon=fm.Icon(color= 'blue', icon='university', prefix="fa")).add_to(m)

m.save('institute_map.html')
m
```
