# Graph Visualization: matplotlib and seaborn

### 📊 Summary Table: `matplotlib` vs `seaborn`

| **Feature**                  | **matplotlib 🧱**                                | **seaborn 🎨**                                 |
|-----------------------------|--------------------------------------------------|------------------------------------------------|
| **Level of control**        | High (more customizable)                         | Medium (auto-adjusts many settings)           |
| **Ease of use**             | Requires more lines of code                      | Simpler and more readable syntax              |
| **Default aesthetics**      | Basic, needs customization                       | Attractive and modern                         |
| **Types of plots**          | Basic (line, bar, pie, etc.)                     | Advanced (violin, swarm, heatmap, etc.)       |
| **Pandas dependency**       | Optional                                         | Works great with DataFrames                   |
| **Fine customization**      | Very detailed (colors, ticks, layout)            | Limited compared to matplotlib                |
| **Complex annotations**     | More flexible for annotations                    | Less focused on annotations                   |
| **Learning curve**          | Slightly steeper                                 | Smoother for beginners                        |
| **Ideal use cases**         | Custom or production-ready visualizations        | Quick and visual exploratory data analysis    |
| **Integration with others** | Seaborn uses matplotlib as its backend           | Built on matplotlib (they complement each other!) |


```python
import matplotlib.pyplot as plt

# Simple line plot
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

plt.plot(x, y)
plt.title("Simple Line Plot")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.show()
```

```python
plt.plot(x, y, color='green', linestyle='--', marker='o', label='Line 1')
plt.title("Customization")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import seaborn as sns
import pandas as pd

# Dataset of example
df = sns.load_dataset("tips")
```

| **Column**      | **Description**                                         |
|-----------------|---------------------------------------------------------|
| `total_bill`    | Total bill (in dollars) for the meal                    |
| `tip`           | Tip given (in dollars)                                  |
| `sex`           | Gender of the customer (`Male` / `Female`)              |
| `smoker`        | Whether the customer was a smoker (`Yes` / `No`)        |
| `day`           | Day of the week (`Thur`, `Fri`, `Sat`, `Sun`)          |
| `time`          | Time of day (`Lunch` / `Dinner`)                        |
| `size`          | Number of people at the table                           |

```python
sns.scatterplot(data=df, x="total_bill", y="tip", hue="sex")
plt.title("Scatterplot with Seaborn")
plt.show()
```

```python
sns.boxplot(x="day", y="total_bill", data=df)
plt.title("Boxplot by Day")
plt.show()
```

```python
sns.histplot(df["total_bill"], kde=True)
plt.title("Histograma con KDE")
plt.show()
```

```python
correlation = df.corr(numeric_only=True)
sns.heatmap(correlation, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
import seaborn as sns
titanic = sns.load_dataset("titanic")
titanic.head()
```

```python
# Gráfico de barras de supervivencia por clase
sns.countplot(x='pclass', hue='survived', data=titanic)
plt.title("Tasa de Supervivencia por Clase")
plt.xlabel("Clase")
plt.ylabel("Número de pasajeros")
plt.show()
```

```python
# Gráfico de distribución de las tarifas con respecto a la supervivencia
sns.boxplot(x='survived', y='fare', data=titanic)
plt.title("Distribución de la Tarifa según Supervivencia")
plt.xlabel("Supervivencia (0 = No, 1 = Sí)")
plt.ylabel("Tarifa")
plt.show()
```

```python
# ¿Cuál es la relación entre edad y supervivencia?
sns.boxplot(x="survived", y="age", data=titanic)
plt.title("Edad vs Supervivencia")
plt.show()
```

```python
# Gráfico de barras de supervivencia por sexo
sns.countplot(x='sex', hue='survived', data=titanic)
plt.title("Supervivencia por Sexo")
plt.xlabel("Sexo")
plt.ylabel("Número de pasajeros")
plt.show()
```
