````python
texto_articulo = """
La transformación digital en América Latina ha acelerado en los últimos años, especialmente tras la pandemia.
Muchos gobiernos y empresas privadas adoptaron nuevas tecnologías para garantizar la continuidad de sus servicios.
La digitalización del Estado, el trabajo remoto, la educación virtual y los pagos electrónicos son algunos de los principales avances observados.
Sin embargo, persisten grandes desafíos relacionados con la conectividad, la ciberseguridad y la capacitación digital de la población.

Por ejemplo, en zonas rurales de países como Perú o Bolivia, el acceso a internet de calidad sigue siendo limitado, lo que afecta directamente a estudiantes y trabajadores remotos.
Además, muchas empresas medianas y pequeñas aún no cuentan con estrategias claras de ciberseguridad, dejándolas expuestas a posibles ataques.

A pesar de estos retos, la región ha mostrado avances notables. Iniciativas de gobiernos para simplificar trámites en línea, plataformas educativas con contenido gratuito, y el crecimiento de servicios fintech han cambiado la forma en que las personas acceden a servicios y productos.

En conclusión, si bien América Latina ha dado pasos importantes hacia la digitalización, aún es fundamental cerrar las brechas de acceso, invertir en infraestructura tecnológica y fortalecer las competencias digitales de la población para que los beneficios sean realmente inclusivos y sostenibles.
"""
````

````python
prompt_resumen = f"""
Resume el siguiente artículo en 5 oraciones claras, en lenguaje sencillo, manteniendo las ideas principales:

\"\"\"
{texto_articulo}
\"\"\"
"""
````

````python
response = client.chat.completions.create(
    model="gpt-4",
    messages=[
        {"role": "system", "content": "Eres un redactor experto en resumir artículos para estudiantes."},
        {"role": "user", "content": prompt_resumen}
    ],
    temperature=0
)

print("📝 Resumen generado:\n")
print(response.choices[0].message.content.strip())
````

````python
def clasificar_tono(texto):
    prompt = f"""
Act as a People Analytics analyst specialized in organizational climate and exit interviews.

Your task is to analyze the following description written by an employee upon leaving the company, and identify the **predominant emotional tone** in their message.

Evaluate the text based on its emotional content, the language used, and the implicit workplace context.

Respond with a single word: Positive, Neutral, or Negative.


Text:
\"\"\"
{texto}
\"\"\"
"""
    try:
        respuesta = client.chat.completions.create(
            model="gpt-4",
            messages=[
                {"role": "system", "content": "You are an expert in emotional analysis of job interviews."},
                {"role": "user", "content": prompt}
            ],
            temperature=0
        )
        return respuesta.choices[0].message.content.strip()
    except Exception as e:
        return f"Error: {e}"
````

````python
def validar_registro_con_gpt(motivo, apto, descripcion):
    prompt = f"""
Analyze the consistency of the following exit interview. Evaluate whether there is a contradiction between the reason for leaving, the description, and the "Eligible for Rehire" field.

Return one of the following responses:
- "Correct" (if everything is consistent)
- "Error: inconsistency detected" (if there is a major contradiction, such as fraud and rehire = Yes)

Primary Reason  : {motivo}
Eligible for Rehire: {apto}

Description:
\"\"\"
{descripcion}
\"\"\"
"""
    try:
        respuesta = client.chat.completions.create(
            model="gpt-4",
            messages=[
                {"role": "system", "content": "You are a Human Resources data auditor who detects inconsistencies in exit interviews."},
                {"role": "user", "content": prompt}
            ],
            temperature=0
        )
        return respuesta.choices[0].message.content.strip()
    except Exception as e:
        return f"Error: {e}"
````

````python
# Clasificación de tono
df_entrevistas["Tono emocional"] = df_entrevistas["Descripción del cese"].apply(clasificar_tono)

# Validación lógica
df_entrevistas["Validación"] = df_entrevistas.apply(
    lambda row: validar_registro_con_gpt(
        row["Motivo primario"],
        row["Apto para reingreso"],
        row["Descripción del cese"]
    ),
    axis=1
)

# Mostrar resultados finales
df_entrevistas
````

# 🧭 Mapping Possibilities in Prompt Engineering

## 💡 Main Categories

| Category                  | Objective                                    | Practical Examples                                                  | Prompt Techniques                      | Level       |
|---------------------------|----------------------------------------------|----------------------------------------------------------------------|----------------------------------------|-------------|
| 🧠 Understanding & Analysis | Understand or interpret content              | Summarization, sentiment analysis, topic classification             | Direct instruction + chunking          | Basic       |
| ✍️ Rewriting & Style       | Rephrase or adapt language                   | Rewrite emails, simplify texts, adjust tone                         | Reformulation + role                   | Basic       |
| 📊 Structured Extraction   | Convert free text into structured data       | JSON from invoices, contracts, CVs, open-ended responses            | Few-shot + delimiters + role           | Intermediate|
| 📚 Knowledge Synthesis     | Combine or summarize multiple sources        | Compare laws, merge opinions, generate conclusions                  | Instruction + reflection               | Intermediate|
| 🧾 Document Automation     | Generate or complete standard formats        | Meeting minutes, reports, contracts, job descriptions               | Template + placeholders                | Intermediate|
| 🧩 Validation & Review     | Detect errors or inconsistencies             | Logical verification in forms, contradictions, fraud detection     | Conditions + logic + critical role     | Intermediate|
| 🧠 Logical Reasoning       | Solve problems and assess scenarios          | Step-by-step explanations, causal inferences                        | Chain-of-thought + expert role         | Advanced    |
| 🤖 Multi-turn Interaction  | Maintain conversational coherence            | Assistants, guided forms, interviews                                | Role + conversational history          | Advanced    |
| 🎨 Assisted Creativity     | Generate new or original content             | Slogans, stories, campaigns, brainstorming                          | Style + inspiration + temperature      | Advanced    |
| 💼 Professional Optimization | Automate specific business tasks          | CV analysis, reports, legal responses, accounting dashboards        | Fine-tuning + RAG + tools              | Advanced    |

## 🔁 Prompt Scalability

1. **Single Prompt**: One task (summarize, classify, validate).
2. **Composite Prompt**: Chain of subtasks (extract + validate + analyze).
3. **Automated Pipeline**: Full workflow (input → processing → output).
4. **Product Integration**: In real systems (HR, ERP, CRM, LMS, etc.).
