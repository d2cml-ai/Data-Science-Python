```python
from google.colab import files

# Load PDF
uploaded = files.upload()

# Detect file name
for filename in uploaded.keys():
    pdf_path = filename
print(f"📁 Archivo cargado: {pdf_path}")
```

```python
import fitz  # PyMuPDF

def extraer_texto_pdf(ruta_pdf):
    doc = fitz.open(ruta_pdf)
    texto = ""
    for pagina in doc:
        texto += pagina.get_text()
    doc.close()
    return texto

cv_texto = extraer_texto_pdf(pdf_path)
print("✅ Texto extraído del CV:")
print(cv_texto[:5000])  # muestra primeros 1000 caracteres
```

```python
from openai import OpenAI
from getpass import getpass

api_key = getpass("🔑 Ingresa tu API key de OpenAI: ").strip()
client = OpenAI(api_key=api_key)
```

```python
prompt = f"""
You are an expert in talent acquisition for tech and financial companies.
Analyze the following résumé and extract:

1. Professional experience (company, role, years)
2. Certifications (name, year, issuing organization if mentioned)
3. Mastered technologies (languages, platforms, tools)
4. Signs of inconsistencies or gaps (unclear dates, unsupported certifications)
5. Overall relevance of the profile for a senior data analyst position

CV:
\"\"\"
{cv_texto}
\"\"\"
"""
```

```python
response = client.chat.completions.create(
    model="gpt-4",
    messages=[
        {"role": "system", "content": "You are a specialist in CV analysis for tech/banking recruitment processes."},
        {"role": "user", "content": prompt}
    ],
    temperature=0
)

print("📋 CV Analysis Report:\n")
print(response.choices[0].message.content.strip())
```

```python
prompt = f"""
You are a talent acquisition expert for tech positions in banking, insurance, or technology sectors.

Your task is to analyze the following CV and compare it with the ideal profile defined below.

You must perform:
1. Extraction of professional experience (company, role, years).
2. Identification of mastered technical tools (SQL, Python, BI, etc.).
3. Relevant certifications (name, year, issuing entity if provided).
4. Evaluation of soft skills and leadership ability.
5. Comparison with the ideal profile.
6. Numerical scoring per category (see criteria).
7. Total score (scale from 0 to 100) and hiring recommendation.

🎯 Use this result template:

----
🔍 **CV Evaluation Summary**

📌 Experience (____ / 30):
- ...

🛠️ Technical Tools (____ / 25):
- ...

📚 Certifications (____ / 15):
- ...

💬 Soft Skills and Leadership (____ / 20):
- ...

🎁 Added or Differentiating Value (____ / 10):
- ...

📊 **Total Score**: ___ / 100
🎯 **Recommendation**: (Yes / No / Requires technical interview)

📝 General Comments:
- ...
----

🔍 Ideal job profile:

{ideal_profile}

👤 Candidate’s CV:
\"\"\"
{cv_texto}
\"\"\"
"""
```

```python
response = client.chat.completions.create(
    model="gpt-4",
    messages=[
        {"role": "system", "content": "You are a talent acquisition expert who compares CVs against ideal profiles."},
        {"role": "user", "content": prompt}
    ],
    temperature=0
)

print("📋 Report with Comparison and Scoring:\n")
print(response.choices[0].message.content.strip())
```

✅ Conclusion: Benefits of this Automation
Automating CV analysis through Prompt Engineering + language models like GPT-4 generates strategic benefits for both the Human Resources department and the business as a whole:

🔍 Consistency: More objective and standardized evaluations.

⏱️ Time savings: Automatic preliminary filtering of large volumes of candidates.

📊 Structured evaluation: Clear scoring in key categories.

🧠 In-depth analysis: Not only extracts information, but also compares, interprets, and recommends.

🎯 Focus on added value: Helps identify candidates aligned with the role’s culture and vision, not just the most "qualified."

This approach takes the selection process to the next level by integrating contextual intelligence and automatic comparative analysis, and it is applicable to various professional, technical, and creative profiles.
