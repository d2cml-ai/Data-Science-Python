```python
import streamlit as st
import pandas as pd
import numpy as np
import datetime
import os
from io import BytesIO

# --- Configuración inicial ---
st.set_page_config(page_title="Encuesta Laboral Interactiva", layout="centered")

# --- Título principal ---
st.title("📋 Encuesta de Satisfacción Laboral")

# --- Texto libre ---
nombre = st.text_input("🧑‍💼 Ingresa tu nombre")

# --- Selección individual ---
genero = st.radio("🧬 Género", ["Masculino", "Femenino", "Otro"])

# --- Selección desplegable ---
departamento = st.selectbox("📍 Selecciona tu departamento", 
                            ["Lima", "Arequipa", "Cusco", "Piura", "La Libertad"])

# --- Multiselección ---
herramientas = st.multiselect("🛠️ Herramientas que usas frecuentemente",
                              ["Excel", "Python", "R", "Power BI", "SQL", "Tableau"])

# --- Slider numérico ---
edad = st.slider("🎂 Edad", min_value=18, max_value=65, value=30)

# --- Número ingresado ---
horas_trabajo = st.number_input("⏳ Horas trabajadas por semana", min_value=0, max_value=80, value=40)

# --- Checkbox opcional ---
mostrar = st.checkbox("📊 Mostrar resumen de tus respuestas")

# --- Date/time ---
fecha_nacimiento = st.date_input("📅 Fecha de nacimiento")
hora_inicio = st.time_input("🕘 Hora de inicio laboral")

# --- Color picker ---
color_favorito = st.color_picker("🎨 Elige tu color favorito")

# --- Subir archivo ---
archivo = st.file_uploader("📁 Sube tu CV (opcional)", type=["pdf", "docx"])

# --- Botón de envío ---
if st.button("✅ Enviar respuestas"):
    respuesta = {
        "nombre": nombre,
        "genero": genero,
        "departamento": departamento,
        "herramientas": ", ".join(herramientas),
        "edad": edad,
        "horas_trabajo": horas_trabajo,
        "fecha_nacimiento": fecha_nacimiento,
        "hora_inicio": str(hora_inicio),
        "color": color_favorito
    }
    df_respuesta = pd.DataFrame([respuesta])
    archivo_existente = os.path.isfile("respuestas.csv")
    df_respuesta.to_csv("respuestas.csv", mode='a', index=False, header=not archivo_existente)
    st.success("✅ Respuesta guardada exitosamente en 'respuestas.csv'")

# --- Mostrar resumen si se selecciona checkbox ---
if mostrar and nombre:
    st.subheader("📝 Resumen de respuestas")
    st.write(f"**Nombre:** {nombre}")
    st.write(f"**Género:** {genero}")
    st.write(f"**Departamento:** {departamento}")
    st.write(f"**Herramientas utilizadas:** {', '.join(herramientas)}")
    st.write(f"**Edad:** {edad} años")
    st.write(f"**Horas trabajadas por semana:** {horas_trabajo}")
    st.write(f"**Fecha de nacimiento:** {fecha_nacimiento}")
    st.write(f"**Hora de inicio:** {hora_inicio}")
    st.write(f"**Color favorito:** {color_favorito}")
    if archivo:
        st.write("**Archivo subido correctamente.**")

# --- Mostrar tabla con respuestas anteriores (si existe el archivo) ---
st.markdown("---")
st.subheader("📈 Respuestas anteriores")

if os.path.isfile("respuestas.csv"):
    df_respuestas = pd.read_csv("respuestas.csv")
    st.dataframe(df_respuestas)

    # --- Botón para descargar en Excel ---
    output = BytesIO()
    with pd.ExcelWriter(output, engine='xlsxwriter') as writer:
        df_respuestas.to_excel(writer, index=False, sheet_name='Respuestas')
        writer.save()
    output.seek(0)

    st.download_button(
        label="📥 Descargar respuestas en Excel",
        data=output,
        file_name="respuestas_encuesta.xlsx",
        mime="application/vnd.openxmlformats-officedocument.spreadsheetml.sheet"
    )
else:
    st.info("Aún no hay respuestas registradas.")
```
