```python
import streamlit as st
import pandas as pd
import numpy as np
import plotly.express as px

# Configuración inicial
st.set_page_config(page_title="Análisis de Encuesta Laboral", layout="wide")

# Simulación del DataFrame
np.random.seed(42)
n = 200
df = pd.DataFrame({
    "id": np.arange(1, n+1),
    "edad": np.random.randint(21, 60, size=n),
    "genero": np.random.choice(["Masculino", "Femenino", "Otro"], size=n, p=[0.45, 0.45, 0.10]),
    "nivel_educativo": np.random.choice(["Secundaria", "Técnica", "Universitaria", "Posgrado"], size=n, p=[0.2, 0.3, 0.35, 0.15]),
    "area_trabajo": np.random.choice(["Administración", "TI", "Ventas", "Producción", "Logística"], size=n),
    "satisfaccion": np.clip(np.random.normal(loc=7, scale=1.5, size=n), 1, 10).round(1),
    "ingreso_mensual": np.random.normal(loc=3000, scale=800, size=n).astype(int),
    "años_experiencia": np.random.randint(0, 35, size=n),
    "desempeño_evaluado": np.random.choice(["Excelente", "Bueno", "Regular", "Malo"], size=n, p=[0.3, 0.45, 0.2, 0.05])
})

st.title("📊 Análisis de Encuesta Laboral")

# Filtros
area = st.selectbox("Selecciona un área de trabajo", df["area_trabajo"].unique())
edad_max = st.slider("Edad máxima", min_value=int(df.edad.min()), max_value=int(df.edad.max()), value=int(df.edad.max()))
mostrar_tabla = st.checkbox("Mostrar tabla de datos filtrados")

# Filtrado
df_filtrado = df[(df.area_trabajo == area) & (df.edad <= edad_max)]

# KPIs
col1, col2 = st.columns(2)
with col1:
    st.metric("💰 Ingreso promedio", f"S/. {df_filtrado.ingreso_mensual.mean():,.2f}")
with col2:
    st.metric("😊 Satisfacción promedio", f"{df_filtrado.satisfaccion.mean():.2f} / 10")

# Visualización
if st.checkbox("Mostrar análisis gráfico"):
    metrica = st.radio("Selecciona una métrica a comparar", ["media", "mediana"])
    st.subheader("🎯 Distribución del desempeño")
    st.bar_chart(df_filtrado["desempeño_evaluado"].value_counts())

    st.subheader("🎓 Nivel educativo")
    st.bar_chart(df_filtrado["nivel_educativo"].value_counts())

    st.subheader("💸 Distribución de ingresos por género")
    fig = px.box(df_filtrado, x="genero", y="ingreso_mensual", color="genero",
                 title="Boxplot de Ingresos por Género")
    st.plotly_chart(fig, use_container_width=True)

# Mostrar tabla
if mostrar_tabla:
    st.subheader("📋 Datos filtrados")
    st.dataframe(df_filtrado)

# Mostrar diccionario de variables
with st.expander("📘 Diccionario de variables"):
    st.markdown("""
    | Variable             | Tipo               | Descripción                                               |
    |----------------------|--------------------|-----------------------------------------------------------|
    | `id`                 | Numérico           | ID del encuestado                                         |
    | `edad`               | Numérico           | Edad del trabajador                                       |
    | `genero`             | Categórica         | Género: M, F, Otro                                        |
    | `nivel_educativo`    | Categórica         | Secundaria, Técnica, Universitaria, Posgrado              |
    | `area_trabajo`       | Categórica         | Administración, TI, Ventas, Producción, etc.              |
    | `satisfaccion`       | Numérico (1-10)    | Nivel de satisfacción general                             |
    | `ingreso_mensual`    | Numérico           | Salario mensual                                           |
    | `años_experiencia`   | Numérico           | Experiencia laboral en años                               |
    | `desempeño_evaluado` | Categórica         | Excelente, Bueno, Regular, Malo                           |
    """)
```
