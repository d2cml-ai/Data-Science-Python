```python
import streamlit as st
import pandas as pd
import numpy as np
import plotly.express as px
```

```python
# Semilla para reproducibilidad
np.random.seed(42)

# Simulamos datos
regiones = ['Lima', 'Arequipa', 'Cusco', 'Piura', 'La Libertad', 'Puno']
anios = list(range(2020, 2025))

data = []

for region in regiones:
    for anio in anios:
        pbi = np.random.normal(8000, 1500)
        pobreza = np.clip(np.random.normal(25, 5), 5, 50)
        desempleo = np.clip(np.random.normal(8, 2), 3, 20)
        educacion = np.random.normal(10, 1.5)
        esperanza = np.random.normal(75, 2)
        desarrollo = np.clip((pbi/10000 + (12 - pobreza)/20 + educacion/15 + esperanza/100)/4, 0, 1)
        
        data.append({
            'region': region,
            'anio': anio,
            'pbi_per_capita': round(pbi, 2),
            'pobreza_pct': round(pobreza, 2),
            'desempleo_pct': round(desempleo, 2),
            'educacion_prom': round(educacion, 1),
            'esperanza_vida': round(esperanza, 1),
            'indice_desarrollo': round(desarrollo, 3)
        })

df = pd.DataFrame(data)
```

```python
import streamlit as st
import pandas as pd
import numpy as np
import plotly.express as px

# ✅ Esta línea debe ser la primera de Streamlit
st.set_page_config(page_title="Dashboard Socioeconómico", layout="wide")

# --- Generación de datos simulados ---
np.random.seed(42)
regiones = ['Lima', 'Arequipa', 'Cusco', 'Piura', 'La Libertad', 'Puno']
anios = list(range(2020, 2025))
data = []

for region in regiones:
    for anio in anios:
        pbi = np.random.normal(8000, 1500)
        pobreza = np.clip(np.random.normal(25, 5), 5, 50)
        desempleo = np.clip(np.random.normal(8, 2), 3, 20)
        educacion = np.random.normal(10, 1.5)
        esperanza = np.random.normal(75, 2)
        desarrollo = np.clip((pbi/10000 + (12 - pobreza)/20 + educacion/15 + esperanza/100)/4, 0, 1)
        
        data.append({
            'region': region,
            'anio': anio,
            'pbi_per_capita': round(pbi, 2),
            'pobreza_pct': round(pobreza, 2),
            'desempleo_pct': round(desempleo, 2),
            'educacion_prom': round(educacion, 1),
            'esperanza_vida': round(esperanza, 1),
            'indice_desarrollo': round(desarrollo, 3)
        })

df = pd.DataFrame(data)

# --- Sidebar de filtros ---
st.sidebar.header("Filtros")
region_sel = st.sidebar.selectbox("Selecciona una región", sorted(df["region"].unique()))
anio_sel = st.sidebar.slider("Selecciona el año", min_value=2020, max_value=2024, value=2022)

df_region = df[df["region"] == region_sel]
df_anio = df_region[df_region["anio"] == anio_sel]

# --- Título principal ---
st.title("📊 Dashboard Socioeconómico por Región")
st.markdown(f"Análisis de indicadores en **{region_sel}** para el año **{anio_sel}**")

# --- KPIs principales ---
col1, col2, col3, col4 = st.columns(4)

if not df_anio.empty:
    col1.metric("💰 PBI per cápita (USD)", f"${df_anio.iloc[0]['pbi_per_capita']:,}")
    col2.metric("📉 % Pobreza", f"{df_anio.iloc[0]['pobreza_pct']}%")
    col3.metric("📈 % Desempleo", f"{df_anio.iloc[0]['desempleo_pct']}%")
    col4.metric("🎓 Educación promedio (años)", df_anio.iloc[0]['educacion_prom'])

# --- Gráficos de evolución ---
st.subheader("📈 Evolución de indicadores en el tiempo")

col5, col6 = st.columns(2)

with col5:
    st.line_chart(df_region.set_index("anio")[["pbi_per_capita", "pobreza_pct"]])

with col6:
    st.bar_chart(df_region.set_index("anio")[["desempleo_pct", "educacion_prom"]])

# --- Gráfico interactivo con Plotly ---
st.subheader("📊 Comparación interactiva")

fig = px.scatter(df_region,
                 x="pbi_per_capita",
                 y="indice_desarrollo",
                 size="educacion_prom",
                 color="anio",
                 hover_name="anio",
                 title=f"Relación entre PBI y Desarrollo en {region_sel}",
                 labels={"pbi_per_capita": "PBI per cápita", "indice_desarrollo": "Índice de Desarrollo"})

st.plotly_chart(fig, use_container_width=True)

# --- Footer ---
st.markdown("---")
st.caption("App creada para clase práctica de Streamlit – ©2025")
```
