# BCRP

## Formato API BRCP

Introduction to the BCRP API

The Central Reserve Bank of Peru (BCRP) API provides access to macroeconomic statistical series such as inflation, GDP, and exchange rates, facilitating economic and financial analysis.


Main Features:

HTTP Methods: GET and POST.

Output Formats: HTML, Graph, XLS, XML, JSON, JSONP, TXT, CSV.

Access to data in Spanish and English.

**URL BASE**

```python
https://estadisticas.bcrp.gob.pe/estadisticas/series/api/{CODIGO_SERIE}/{FORMATO}/{FECHA_INICIO}/{FECHA_FIN}
```

Main Parameters:

COD_SERIE: Code of the statistical series.

FORMATO: Output format (json, xml, csv, xls, etc.).

FECHA_INI/FECHA_FIN: Date range (YYYY-MM-DD).

IDIOMA: Spanish (esp) or English (ing).

HTTP Methods: GET and POST
GET:

It is the most commonly used method in public APIs.

It allows requesting data from a server without modifying the stored information.



| **Feature**           | **GET**                               | **POST**                               |
|-----------------------|---------------------------------------|----------------------------------------|
| **Purpose**           | Retrieve data                        | Send data to the server               |
| **Visibility**        | Parameters in the URL                | Parameters in the request body        |
| **Security**          | Not secure for sensitive data        | More secure for sensitive data        |
| **Typical Use**       | Retrieve information (searches, etc.)| Submit forms, upload files            |
| **Data Size**         | Limited by the length of the URL     | Can handle large amounts of data      |



### GET

```python
import requests

# Example URL
url = "https://jsonplaceholder.typicode.com/posts"

# Making a GET request
response = requests.get(url)

# Checking the response status
if response.status_code == 200:
    print("GET request successful")
    # Print the first 3 items from the response
    posts = response.json()  # Parse JSON
    print(posts[:3])  # Show the first 3 posts
else:
    print("Error in GET request:", response.status_code)

```

```python
def get_exchange_rate(series_code):
    url = f"https://estadisticas.bcrp.gob.pe/estadisticas/series/api/{series_code}/json"
    response = requests.get(url)

    # Check if the response is valid
    if response.status_code != 200:
        print(f"Error: Code {response.status_code} - Data could not be retrieved")
        return None
    
    try:
        data = response.json()
    except requests.exceptions.JSONDecodeError:
        print("Error: The response is not in JSON format")
        print(response.text)  # Show response content for debugging
        return None

    # Extract data from the series
    values = data.get('periods', [])  # Avoid KeyError
    dates = [item['name'] for item in values]
    exchange_rate_values = [float(item['values'][0]) for item in values]

    return pd.DataFrame({'Date': dates, 'TC': exchange_rate_values})

df_purchase = get_exchange_rate("PN01205PM")
df_sale = get_exchange_rate("PN01206PM")
```

```python
# Configurar Selenium (ajusta la ruta del driver)
import time
driver = webdriver.Chrome()
driver.maximize_window()


# URL de la página que contiene los datos
url = "https://estadisticas.bcrp.gob.pe/estadisticas/series/mensuales/"  
driver.get(url)

# Esperar carga completa (puedes aumentar el tiempo si la web es lenta)
time.sleep(5)

# 🔹 Extraer HTML de la página (IMPORTANTE)
html = driver.page_source

# Cerrar el navegador
#driver.quit()

# Verificar que HTML se ha obtenido correctamente
if not html:
    raise ValueError("❌ Error: No se pudo obtener el HTML de la página.")

# Procesar con BeautifulSoup
soup = BeautifulSoup(html, 'html.parser')

# Extraer todos los enlaces con el patrón esperado
links = soup.find_all("a", href=lambda href: href and "/estadisticas/series/mensuales/resultados/" in href)

# 🔹 Extraer todas las categorías
categorias = soup.find_all("h2", class_="categoria_titulo")

# Lista para almacenar los resultados
resultados = []

# Iterar por cada categoría
for categoria in categorias:
    # Extraer el título de la categoría (quitando el contenido adicional)
    titulo_categoria = categoria.text.strip().split("\n")[0]  # Tomamos solo la primera línea del título

    # Buscar la siguiente tabla dentro de la categoría
    tabla = categoria.find_next("table", class_="series")
    
    # Si hay una tabla, extraemos los códigos y descripciones
    if tabla:
        filas = tabla.find_all("tr")[1:]  # Ignorar la cabecera
        for fila in filas:
            columnas = fila.find_all("td")
            if len(columnas) >= 3:  # Aseguramos que haya suficientes columnas
                codigo = columnas[1].text.strip()
                descripcion = columnas[2].text.strip()
                resultados.append((titulo_categoria, codigo, descripcion))
```

```python
import requests
import pandas as pd

class BCRPAPI:
    BASE_URL = "https://estadisticas.bcrp.gob.pe/estadisticas/series/api/"
    
    def __init__(self, series_code, start_date=None, end_date=None):
        self.series_code = series_code
        self.start_date = start_date
        self.end_date = end_date
    
    def get_data(self):
        # Si se proporcionan fechas, incluirlas en la URL
        if self.start_date and self.end_date:
            url = f"{self.BASE_URL}{self.series_code}/json/{self.start_date}/{self.end_date}"
        else:
            url = f"{self.BASE_URL}{self.series_code}/json/"
        
        response = requests.get(url)
        
        if response.status_code == 200:
            try:
                data = response.json()
            except requests.exceptions.JSONDecodeError:
                print("Error: La respuesta de la API no es un JSON válido.")
                print("Contenido de la respuesta:", response.text)
                return None
            
            if "periods" in data and isinstance(data["periods"], list):
                fechas = [item["name"] for item in data["periods"]]
                valores = [float(item["values"][0]) if item["values"] else None for item in data["periods"]]

                # Convertir fechas a un formato estándar (año-mes) para asegurar orden correcto
                df = pd.DataFrame({"Fecha": fechas, "Valor": valores})
                df["Fecha"] = pd.to_datetime(df["Fecha"], format="%b.%Y", errors="coerce")
                df = df.dropna(subset=["Fecha"])  # Eliminar valores inválidos
                df = df.sort_values(by="Fecha", ascending=True).reset_index(drop=True)  # Ordenar por fecha
                return df
            else:
                print("No se encontraron datos para la serie.")
                return None
        else:
            print("Error al obtener datos, código:", response.status_code)
            print("Respuesta:", response.text)
            return None

# Ejemplo de uso sin rango de fechas para obtener toda la data posible
bcrp = BCRPAPI("PN00026MM")
data = bcrp.get_data()
if data is not None:
    print(data)
```

```python
import requests
import pandas as pd
from datetime import datetime

class BCRPAPI:
    BASE_URL = "https://estadisticas.bcrp.gob.pe/estadisticas/series/api/"
    
    def __init__(self, series_code, start_year=1994):
        self.series_code = series_code
        self.start_year = start_year
        self.end_year = datetime.now().year  # Año actual
    
    def get_data(self):
        all_data = []
        
        for year in range(self.start_year, self.end_year + 1):
            url = f"{self.BASE_URL}{self.series_code}/json/{year}-1/{year}-12"
            response = requests.get(url)
            
            if response.status_code == 200:
                try:
                    data = response.json()
                except requests.exceptions.JSONDecodeError:
                    print(f"Error: JSON inválido para el año {year}.")
                    continue
                
                if "periods" in data and isinstance(data["periods"], list):
                    fechas = [item["name"] for item in data["periods"]]
                    valores = [float(item["values"][0]) if item["values"] else None for item in data["periods"]]
                    df = pd.DataFrame({"Fecha": fechas, "Valor": valores})
                    all_data.append(df)
            else:
                print(f"Error al obtener datos para el año {year}, código:", response.status_code)
        
        if all_data:
            final_df = pd.concat(all_data, ignore_index=True)
            final_df["Fecha"] = pd.to_datetime(final_df["Fecha"], format="%b.%Y", errors="coerce")
            final_df = final_df.dropna(subset=["Fecha"])  # Eliminar valores inválidos
            final_df = final_df.sort_values(by="Fecha", ascending=True).reset_index(drop=True)
            return final_df
        else:
            print("No se encontraron datos.")
            return None

# Ejemplo de uso: Obtener toda la data desde 1994
bcrp = BCRPAPI("PN00026MM")
data = bcrp.get_data()
if data is not None:
    print(data)
```

| **Characteristic**        | **Class**                                           | **Function**                                        |
|---------------------------|-----------------------------------------------------|----------------------------------------------------|
| **Definition**            | Defined using the `class` keyword.                  | Defined using the `def` keyword.                   |
| **Purpose**               | Used to create objects and encapsulate data and methods. | Used to execute a block of code.                   |
| **Main Use**              | Model complex entities and behaviors.              | Perform a specific task or calculation.            |
| **Instantiation**         | An instance (object) of the class is created.      | Called directly to execute its code.               |
| **Methods**               | Can contain multiple methods (functions) that operate on the class's data. | Does not contain methods inside itself.            |
| **Scope**                 | Has attributes that can be accessed from different methods within the class. | Variables defined within a function are only accessible inside it. |
| **Example**               | `class Person:`<br>`    def __init__(self, name):`<br>`        self.name = name` | `def greet():`<br>`    print("Hello World")`       |
| **Advantages**            | Allows code reuse and organization.                | Simple and quick for small tasks.                  |
| **Return**                | Can return values, but is not limited to that.     | Returns values using `return`.                     |
| **State**                 | Can maintain state through attributes.             | Does not maintain state between executions.        |


```python
import requests
import pandas as pd

class BCRPAPI:
    BASE_URL = "https://estadisticas.bcrp.gob.pe/estadisticas/series/api/"
    
    def __init__(self, series_code, start_date=None, end_date=None):
        self.series_code = series_code
        self.start_date = start_date
        self.end_date = end_date
    
    def get_data(self):
        # Construcción de la URL con rango de fechas si se proporciona
        if self.start_date and self.end_date:
            url = f"{self.BASE_URL}{self.series_code}/json/{self.start_date}/{self.end_date}"
        else:
            url = f"{self.BASE_URL}{self.series_code}/json/"
        
        response = requests.get(url)
        
        if response.status_code == 200:
            try:
                data = response.json()
            except requests.exceptions.JSONDecodeError:
                print("Error: La respuesta de la API no es un JSON válido.")
                print("Contenido de la respuesta:", response.text)
                return None
            
            if "periods" in data and isinstance(data["periods"], list):
                fechas = [item["name"] for item in data["periods"]]
                valores = [float(item["values"][0]) if item["values"] else None for item in data["periods"]]
                df = pd.DataFrame({"Fecha": fechas, "Valor": valores})
                df = df.sort_values(by="Fecha", ascending=True).reset_index(drop=True)  # Ordenar por fecha
                return df
            else:
                print("No se encontraron datos para la serie.")
                return None
        else:
            print("Error al obtener datos, código:", response.status_code)
            print("Respuesta:", response.text)
            return None

# Ejemplo de uso con RD38099BM y rango de fechas especificado
bcrp = BCRPAPI("RD38099BM", "2000-7", "2025-2")
data = bcrp.get_data()
if data is not None:
    print(data)
```

# WORLD BANK

```python
import requests
import pandas as pd

class WorldBankAPI:
    """
    Clase para interactuar con la API del Banco Mundial.
    Permite buscar indicadores y obtener datos económicos por país.
    """
    
    BASE_URL = "https://api.worldbank.org/v2"
    
    def __init__(self, format_type="json", per_page=1000):
        """
        Inicializa la API con formato de datos y cantidad de resultados por página.
        """
        self.format_type = format_type
        self.per_page = per_page

    def search_indicators(self, search_term):
        """
        Busca indicadores en la API del Banco Mundial según un término de búsqueda.
        :param search_term: (str) Palabra clave del indicador (Ej: "GDP", "Inflation")
        :return: DataFrame con los indicadores encontrados.
        """
        url = f"{self.BASE_URL}/indicator?format={self.format_type}&per_page={self.per_page}"
        response = requests.get(url)

        if response.status_code == 200:
            data = response.json()[1]  # La data está en la segunda posición del JSON
            
            # Filtrar indicadores que contengan el término de búsqueda en su nombre
            filtered_data = [item for item in data if search_term.lower() in item["name"].lower()]
            
            # Convertir a DataFrame
            df = pd.DataFrame(filtered_data)
            df = df[["id", "name", "unit", "source"]]
            return df
        else:
            print("Error en la solicitud:", response.status_code)
            return None

    def get_indicator_data(self, indicator_id, country_code="ALL", start_year=2000, end_year=2024):
        """
        Obtiene datos históricos de un indicador para un país o todos los países.
        :param indicator_id: (str) Código del indicador (Ej: "NY.GDP.MKTP.CD" para PIB)
        :param country_code: (str) Código del país (Ej: "PE" para Perú, "ALL" para todos)
        :param start_year: (int) Año de inicio
        :param end_year: (int) Año de fin
        :return: DataFrame con los datos del indicador
        """
        url = f"{self.BASE_URL}/country/{country_code}/indicator/{indicator_id}?date={start_year}:{end_year}&format={self.format_type}&per_page={self.per_page}"
        response = requests.get(url)

        if response.status_code == 200:
            data = response.json()[1]  # La data está en la segunda posición del JSON
            
            # Convertir a DataFrame
            df = pd.DataFrame(data)
            df = df[["country", "date", "value"]]
            
            # Extraer solo el nombre del país
            df["country"] = df["country"].apply(lambda x: x["value"])
            
            return df
        else:
            print("Error en la solicitud:", response.status_code)
            return None

    def export_to_excel(self, df, filename="world_bank_data.xlsx"):
        """
        Exporta un DataFrame a un archivo Excel.
        :param df: (DataFrame) Datos a exportar.
        :param filename: (str) Nombre del archivo.
        """
        if df is not None:
            df.to_excel(filename, index=False)
            print(f"Datos exportados a {filename}")
        else:
            print("No hay datos para exportar.")

# ========== EJEMPLO DE USO ==========
if __name__ == "__main__":
    wb_api = WorldBankAPI()

    # Buscar indicadores que contienen "GDP"
    indicadores_gdp = wb_api.search_indicators("GDP")
    print(indicadores_gdp.head())

    # Obtener datos del PIB de Perú desde el 2000 hasta 2024
    data_pib_peru = wb_api.get_indicator_data("NY.GDP.MKTP.CD", "PE", 2000, 2024)
    print(data_pib_peru.head())

    # Exportar los datos a un archivo Excel
    wb_api.export_to_excel(data_pib_peru, "PIB_Peru.xlsx")
```
