```python
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.by import By
import time

# Leer los links desde un archivo csv
df = pd.read_csv("links.csv", header=None, names=["URL"])
links_empleos = df["URL"].tolist()

# Configurar Selenium
driver = webdriver.Chrome()
detalles_empleos = []

for link in links_empleos:
    driver.get(link)
    time.sleep(3)

    try:
        # Extraigo los datos
        titulo_empleo = driver.find_element(By.XPATH, "//h1").text
        descripcion = driver.find_elements(By.XPATH, "//p[starts-with(@class, 'sc-')]") # Busco todos los párrafos con clases que empiezan con 'sc-'
        
        distrito = driver.find_element(By.XPATH, "//h2[contains(text(), 'Peru')]").text
        modalidad = driver.find_element(By.XPATH, "//p[contains(text(), 'Remoto') or contains(text(), 'Híbrido') or contains(text(), 'Presencial')]").text

 # Guardo en la lista
        detalles_empleos.append({
            "Título del empleo": titulo_empleo,
            "Descripción": descripcion,
            "Distrito": distrito,
            "Modalidad": modalidad
        })

        print(f" Extraído: {titulo_empleo}")

    except Exception as e:
        print(f" Error en {link}: {e}")

# Cerrar navegador
driver.quit()

# Guardar en CSV en el formato solicitado
df_resultados = pd.DataFrame(detalles_empleos, columns=["Título del empleo", "Descripcion", "Distrito", "Modalidad"])
df_resultados.to_csv("empleos_extraidos.csv", index=False)

print("\n Archivo 'empleos_extraidos.csv' guardado con éxito.")
```
