## Codigo 

```python
import requests  
import pandas as pd  
from datetime import datetime  

# Función para obtener concursos  
def get_contests():  
    """Obtiene concursos de Codeforces entre julio y diciembre de 2024."""  
    url = "https://codeforces.com/api/contest.list"  
    response = requests.get(url)  
    contests = []  

    if response.status_code == 200:  
        data = response.json()  
        if data['status'] == 'OK':  
            for contest in data['result']:  
                start_time = datetime.fromtimestamp(contest['startTimeSeconds'])  
                if start_time >= datetime(2024, 7, 1) and start_time <= datetime(2024, 12, 31):  
                    contests.append({  
                        'id_concurso': contest['id'],  # Aquí 'id' es correcto  
                        'nombre_concurso': contest['name'],  
                        'hora_inicio_concurso': start_time,  
                        'duracion': contest['durationSeconds'],  
                        'tipo': contest['type'],  
                        'fase': contest['phase']  
                    })  
    return contests  

# Función para obtener los problemas de un concurso  
def get_problems(contest_id):  
    """Obtiene problemas de un concurso específico."""  
    url = f"https://codeforces.com/api/contest.standings?contestId={contest_id}&from=1&count=1"  
    response = requests.get(url)  
    problems = []  

    if response.status_code == 200:  
        data = response.json()  
        if data['status'] == 'OK':  
            for problem in data['result']['problems']:  
                problems.append({  
                    'problem_name': problem['name'],  
                    'rating': problem.get('rating', 'N/A')  # Calificación del problema  
                })  
    return problems  

# Función para obtener envíos de un usuario específico  
def get_user_submissions(handle, contest_id):  
    """Obtiene los envíos de un usuario para un concurso específico."""  
    url = f"https://codeforces.com/api/user.status?handle={handle}&contestId={contest_id}"  
    response = requests.get(url)  
    submissions = []  

    if response.status_code == 200:  
        data = response.json()  
        if data['status'] == 'OK':  
            for submission in data['result']:  
                submissions.append({  
                    'problem_name': submission['problem']['name'],  
                    'verdict': submission['verdict'],  
                    'namespace': submission['problem']['contestId'],  
                    'runtime': submission['timeConsumedMillis'] // 1000,  # segundos  
                    'memory': submission['memoryConsumedBytes'] // 1024,  # KB  
                    'language': submission.get('programmingLanguage', 'N/A')  # <--- aquí agregamos el lenguaje  
                })  
    return submissions 

# Main   
if __name__ == "__main__":  
    contests = get_contests()  
    user_handle = 'tourist'  # Cambiar al handle del usuario deseado  

    all_data = []  
    problem_ratings = {}  # Diccionario para almacenar las calificaciones de los problemas  

    # Obtenemos problemas y sus ratings para cada concurso  
    for contest in contests:  
        problems_data = get_problems(contest['id_concurso'])  # Cambiar a contest['id_concurso'] a contest['id']  
        for problem in problems_data:  # Para cada problema, guardamos su rating en el diccionario  
            problem_ratings[problem['problem_name']] = problem['rating']  

        submissions = get_user_submissions(user_handle, contest['id_concurso'])  
        
        for submission in submissions:  
            all_data.append({  
                    'id_autor': user_handle,  
                    'verdict': submission['verdict'],  
                    'problem_name': submission['problem_name'],  
                    'runtime': submission['runtime'],  
                    'memory': submission['memory'],  
                    'language': submission['language'],  # <--- aquí lo incluyes  
                    'id_concurso': contest['id_concurso'],  
                    'nombre_concurso': contest['nombre_concurso'],  
                    'hora_inicio_concurso': contest['hora_inicio_concurso'],  
                    'rating': problem_ratings.get(submission['problem_name'], 'N/A')  
                })  

    # Crear DataFrame  
    df = pd.DataFrame(all_data)  
    print(df.head())  # Muestra las primeras filas del DataFrame  
```

## Descripción del Conjunto de Datos  

### Puntos Finales de API Utilizados  

Para la obtención de datos, se han utilizado los siguientes puntos finales de la API de Codeforces:  

1. **`contest.list`**: Este punto final se utiliza para obtener una lista de todos los concursos disponibles. En nuestro caso, se filtraron los concursos que ocurren entre julio y diciembre de 2024.  
   - URL: `https://codeforces.com/api/contest.list`  

2. **`contest.standings`**: Este punto final proporciona información sobre los problemas de un concurso específico. Extraemos la lista de problemas y su respectiva calificación.  
   - URL: `https://codeforces.com/api/contest.standings?contestId={contest_id}&from=1&count=1`  

3. **`user.status`**: Este punto final permite obtener el historial de envíos de un usuario específico para un concurso determinado. Extraemos información sobre cada envío, como el veredicto, tiempo de ejecución y uso de memoria.  
   - URL: `https://codeforces.com/api/user.status?handle={handle}&contestId={contest_id}`  

### Estructura del Conjunto de Datos  

El conjunto de datos se ha estructurado como un DataFrame de pandas en el que cada fila representa un envío de un problema realizado por un usuario en un concurso específico. Las columnas incluyen información sobre el autor, el problema, el resultado del envío, entre otros.  

### Variables Analizadas  

A continuación se describen las variables que se han incluido en el conjunto de datos:  

- **`id_autor`**: (str) El identificador del usuario (handle) que realizó el envío. En este caso, se ha utilizado el handle de un usuario específico ("tourist").  

- **`verdict`**: (str) El resultado del envío. Puede ser "OK" si el envío fue aceptado, o un mensaje de error si no lo fue (por ejemplo, "Wrong Answer", "Time Limit Exceeded", etc.).  

- **`problem_name`**: (str) El nombre del problema al que se refiere el envío.  

- **`runtime`**: (int) El tiempo que tomó ejecutar la solución (en segundos). Este valor se obtiene dividiendo `timeConsumedMillis` por 1000.  

- **`memory`**: (int) La cantidad de memoria utilizada por el programa (en kilobytes, KB). Este valor se obtiene dividiendo `memoryConsumedBytes` por 1024.  

- **`id_concurso`**: (int) El identificador único del concurso en el que se llevó a cabo el envío.  

- **`nombre_concurso`**: (str) El nombre del concurso que se corresponde con el `id_concurso`.  

- **`hora_inicio_concurso`**: (datetime) La fecha y hora de inicio del concurso, extraída del objeto de concurso original.  

Este conjunto de datos proporciona una vista comprensiva sobre la actividad de un usuario específico en varios concursos de Codeforces, permitiendo analizar su desempeño en problemas individuales y en los concursos en general.  

## Limpieza y Transformación de Datos

```python
# 1. Manejar campos faltantes  
# Reemplazar valores faltantes en 'rating' por 'N/A' y en 'verdict' por 'Sin resultado'  
df['rating'].fillna('N/A', inplace=True)  
df['verdict'].fillna('Sin resultado', inplace=True)  

# 2. Convertir marcas de tiempo  
# Asegurarnos que 'hora_inicio_concurso' esté en formato datetime si no lo estuviera  
df['hora_inicio_concurso'] = pd.to_datetime(df['hora_inicio_concurso'])  

# 3. Crear variables derivadas  
# Convertir 'memory' de KB a MB  
df['memory_MB'] = df['memory'] / 1024  # Cambia de KB a MB  

# Crear una columna que clasifique el veredicto  
# Podríamos clasificarlo como 'Aceptado' si el veredicto es 'OK'  
# y como 'Rechazado' si es cualquier otro valor  
df['resultado_clasificacion'] = df['verdict'].apply(lambda x: 'Aceptado' if x == 'OK' else 'Rechazado')  

# Si deseas mantener el DataFrame limpio  
df.drop(columns=['memory'], inplace=True)  # Puedes eliminar la columna original si no es necesaria  

# Mostrar información del DataFrame después de la limpieza y transformación  
print(df.info())  
print(df.head()) 
```

## 1. Histograma de tiempos de envío

# 2. Figura del kernel de densidad para las calificaciones máximas y mínimas de los usuarios

```python
import matplotlib.pyplot as plt  
import seaborn as sns  
import pandas as pd  

# Asegúrate de que df['rating'] está en formato numérico, trata de convertir  
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')  # Convierte a numérico y coloca NaN si no puede  

# Eliminar filas donde rating es NaN  
df = df.dropna(subset=['rating'])  

# Graficar la densidad utilizando el nuevo argumento fill  
plt.figure(figsize=(10, 6))  
sns.kdeplot(df['rating'], fill=True)  # Usa fill=True en lugar de shade=True  
plt.title('Densidad de Calificaciones')  
plt.xlabel('Calificación')  
plt.ylabel('Densidad')  
plt.grid()  
plt.show() 
```

gráfica de densidad (KDE) para todas las calificaciones que hay en la columna rating del DataFrame df

los usuarios no dan calificaciones de forma uniforme. Puede haber varios "grupos" de usuarios que tienden a calificar en ciertos rangos. Por ejemplo:

Algunos usuarios dan calificaciones bajas consistentemente (800–1000).

Otros dan calificaciones medias (1800–2000).

Otros dan calificaciones altas (3000+).
